```python
# Reference file used for filtering
guide_file = "HT1-1_full_guide_sequences.txt"

# List ALL files that need to have corresponding lines removed
files = [
    "HT1-1_full_guide_sequences.txt",

    # Add your other files here, for example:
    "HT1-1_indel_frequency_value_percentage.txt",
    "HT1-1_target_sequences_noPAM.txt",
    "HT1-1_target_sequences_with_noise_rc.txt",
]

# ------------------------------------------------------------
# Step 1: Read guide sequences
# ------------------------------------------------------------
with open(guide_file, "r") as f:
    guide_lines = f.readlines()

print(f"Original number of samples: {len(guide_lines)}")

# ------------------------------------------------------------
# Step 2: Identify lines to KEEP
# ------------------------------------------------------------
keep_indices = []

for i, line in enumerate(guide_lines):

    sequence = line.strip()

    # Only examine the first 20 symbols
    last_20 = sequence[-20:]

    # Remove if 4 or more consecutive U's occur
    # Checking "UUUU" is sufficient because UUUUU, UUUUUU, etc.
    # also contain "UUUU".
    if "UUUU" not in last_20:
        keep_indices.append(i)

# ------------------------------------------------------------
# Step 3: Report filtering results
# ------------------------------------------------------------
n_original = len(guide_lines)
n_kept = len(keep_indices)
n_removed = n_original - n_kept

print(f"Removed samples: {n_removed}")
print(f"Remaining samples: {n_kept}")

# ------------------------------------------------------------
# Step 4: Apply identical filtering to every file
# ------------------------------------------------------------
for filename in files:

    with open(filename, "r") as f:
        lines = f.readlines()

    # Safety check: all files must have the same number of lines
    if len(lines) != n_original:
        raise ValueError(
            f"{filename} has {len(lines)} lines, "
            f"but {guide_file} has {n_original} lines."
        )

    # Keep exactly the same sample indices
    filtered_lines = [lines[i] for i in keep_indices]

    # Overwrite the original file
    with open(filename, "w") as f:
        f.writelines(filtered_lines)

    print(
        f"{filename}: {len(lines)} -> "
        f"{len(filtered_lines)} lines"
    )

print("\nFiltering completed successfully.")
```
