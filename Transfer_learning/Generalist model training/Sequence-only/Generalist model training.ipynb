```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
import numpy as np

from torch.utils.data import Dataset, DataLoader, Subset
from sklearn.model_selection import train_test_split
from scipy.stats import spearmanr
import optuna

import copy

# PyG
from torch_geometric.data import Data, Batch
from torch_geometric.nn import GCNConv, global_mean_pool

# NUPACK
from nupack import Model, pairs

np.set_printoptions(threshold=np.inf)
print("Using PyTorch and PyG.")

########################################
# 1. Data Loading
########################################

def load_branch1_data_HTCas9():
    """
    Loads (60,4) data from 'Feature_guide_baseonly_PAM_complete_HTCas9_unique_reduced_seq_feature.txt'.
    """
    data_branch1 = []
    current_array = []
    with open('Feature_guide_baseonly_PAM_complete_HTCas9_unique_reduced_seq_feature.txt', 'r') as f:
        for line in f:
            line = line.strip()
            if line == "":
                if current_array:
                    data_branch1.append(current_array)
                    current_array = []
            else:
                line = line.replace('[','').replace(']', '')
                row = [float(x.strip().replace(',', '')) for x in line.split()]
                current_array.append(row)
    if current_array:
        data_branch1.append(current_array)
    X_branch1 = np.array(data_branch1)
    X_branch1 = X_branch1.reshape(len(X_branch1), 60, 4)
    return X_branch1

def load_branch1_data_HT11():
    """
    Loads (60,4) data from 'Feature_guide_baseonly_PAM_complete_HT11_TTTV_filtered_unique_reduced_seq_feature.txt'.
    """
    data_branch1 = []
    current_array = []
    with open('Feature_guide_baseonly_PAM_complete_HT11_TTTV_filtered_unique_reduced_seq_feature.txt', 'r') as f:
        for line in f:
            line = line.strip()
            if line == "":
                if current_array:
                    data_branch1.append(current_array)
                    current_array = []
            else:
                line = line.replace('[','').replace(']', '')
                row = [float(x.strip().replace(',', '')) for x in line.split()]
                current_array.append(row)
    if current_array:
        data_branch1.append(current_array)
    X_branch1 = np.array(data_branch1)
    X_branch1 = X_branch1.reshape(len(X_branch1), 60, 4)
    return X_branch1

def load_branch1_data_mmc4():
    """
    Loads (60,4) data from 'Feature_guide_baseonly_PAM_complete_RfxCas13d_NM_002794_4_reduced_seq_feature.txt'.
    """
    data_branch1 = []
    current_array = []
    with open('Feature_guide_baseonly_PAM_complete_RfxCas13d_NM_002794_4_reduced_seq_feature.txt', 'r') as f:
        for line in f:
            line = line.strip()
            if line == "":
                if current_array:
                    data_branch1.append(current_array)
                    current_array = []
            else:
                line = line.replace('[','').replace(']', '')
                row = [float(x.strip().replace(',', '')) for x in line.split()]
                current_array.append(row)
    if current_array:
        data_branch1.append(current_array)
    X_branch1 = np.array(data_branch1)
    X_branch1 = X_branch1.reshape(len(X_branch1), 60, 4)
    return X_branch1

def load_branch1_data_mmc5():
    """
    Loads (60,4) data from 'Feature_guide_baseonly_PAM_complete_RfxCas13d_validation_reduced_seq_feature.txt'.
    """
    data_branch1 = []
    current_array = []
    with open('Feature_guide_baseonly_PAM_complete_RfxCas13d_validation_reduced_seq_feature.txt', 'r') as f:
        for line in f:
            line = line.strip()
            if line == "":
                if current_array:
                    data_branch1.append(current_array)
                    current_array = []
            else:
                line = line.replace('[','').replace(']', '')
                row = [float(x.strip().replace(',', '')) for x in line.split()]
                current_array.append(row)
    if current_array:
        data_branch1.append(current_array)
    X_branch1 = np.array(data_branch1)
    X_branch1 = X_branch1.reshape(len(X_branch1), 60, 4)
    return X_branch1

def load_branch1_data_change_seq():
    """
    Loads (60,4) data from 'Feature_guide_baseonly_PAM_complete_change_seq_sampled_2_unique_reduced_seq_feature.txt'.
    """
    data_branch1 = []
    current_array = []
    with open('Feature_guide_baseonly_PAM_complete_change_seq_sampled_2_unique_reduced_seq_feature.txt', 'r') as f:
        for line in f:
            line = line.strip()
            if line == "":
                if current_array:
                    data_branch1.append(current_array)
                    current_array = []
            else:
                line = line.replace('[','').replace(']', '')
                row = [float(x.strip().replace(',', '')) for x in line.split()]
                current_array.append(row)
    if current_array:
        data_branch1.append(current_array)
    X_branch1 = np.array(data_branch1)
    X_branch1 = X_branch1.reshape(len(X_branch1), 60, 4)
    return X_branch1

def load_branch1_data_tiger():
    """
    Loads (60,4) data from 'Feature_guide_baseonly_PAM_complete_TIGER_RfxCas13a_off_targets_EIF3B_filtered_sampled_1_reduced_seq_feature.txt'.
    """
    data_branch1 = []
    current_array = []
    with open('Feature_guide_baseonly_PAM_complete_TIGER_RfxCas13a_off_targets_EIF3B_filtered_sampled_1_reduced_seq_feature.txt', 'r') as f:
        for line in f:
            line = line.strip()

            if line == "":
                if current_array:
                    data_branch1.append(current_array)
                    current_array = []
            else:
                line = line.replace('[','').replace(']', '')
                row = [float(x.strip().replace(',', '')) for x in line.split()]
                current_array.append(row)
    if current_array:
        data_branch1.append(current_array)
    X_branch1 = np.array(data_branch1)
    X_branch1 = X_branch1.reshape(len(X_branch1), 60, 4)
    return X_branch1

def load_branch1_data_iMeta():
    """
    Loads (60,4) data from 'Feature_guide_baseonly_PAM_complete_iMeta_TTTV_filtered_reduced_seq_feature.txt'.
    """
    data_branch1 = []
    current_array = []
    with open('Feature_guide_baseonly_PAM_complete_iMeta_TTTV_filtered_reduced_seq_feature.txt', 'r') as f:
        for line in f:
            line = line.strip()
            if line == "":
                if current_array:
                    data_branch1.append(current_array)
                    current_array = []
            else:
                line = line.replace('[','').replace(']', '')
                row = [float(x.strip().replace(',', '')) for x in line.split()]
                current_array.append(row)
    if current_array:
        data_branch1.append(current_array)
    X_branch1 = np.array(data_branch1)
    X_branch1 = X_branch1.reshape(len(X_branch1), 60, 4)
    return X_branch1

def load_reaction_rates_HTCas9():
    with open('HTCas9_indel_frequency_value_percentage_unique.txt', 'r') as f:
        rates = [float(line.strip()) for line in f.readlines()]
    return np.array(rates)

def load_reaction_rates_HT11():
    with open('HT1-1_indel_frequency_value_percentage_TTTV_filtered.txt', 'r') as f:
        rates = [float(line.strip()) for line in f.readlines()]
    return np.array(rates)

def load_reaction_rates_mmc4():
    with open('filtered_ratio_NM_002794_4.txt', 'r') as f:
        rates = [float(line.strip()) for line in f.readlines()]
    return np.array(rates)

def load_reaction_rates_mmc5():
    with open('RfxCas13d_validation_expression_level_combined.txt', 'r') as f:
        rates = [float(line.strip()) for line in f.readlines()]
    return np.array(rates)

def load_reaction_rates_change_seq():
    with open('normalized_change_seq_log_transformed_20nt_sampled_2.txt', 'r') as f:
        rates = [float(line.strip()) for line in f.readlines()]
    return np.array(rates)

def load_reaction_rates_tiger():
    with open('normalized_off_target_log2_fold_change_d15_values_EIF3B_filtered_sampled_1.txt', 'r') as f:
        rates = [float(line.strip()) for line in f.readlines()]
    return np.array(rates)

def load_reaction_rates_iMeta():
    with open('normalized_iMeta_trans_LbCas12a_sorted_data_30min_TTTV_filtered.txt', 'r') as f:
        rates = [float(line.strip()) for line in f.readlines()]
    return np.array(rates)

# NUPACK models
from nupack import Model, pairs
my_model_RNA = Model(material='rna', ensemble='nostacking', celsius=37, sodium=0.1, magnesium=0.01)
my_model_DNA = Model(material='dna', ensemble='nostacking', celsius=37, sodium=0.1, magnesium=0.01)

def DNA_reverse_complement(DNA):
    complement = {'A': 'T', 'C': 'G', 'G': 'C', 'T': 'A'}
    return ''.join(complement.get(base, base) for base in reversed(DNA))

########################################
# 3. CNN, GNN, MLP Branches (One hidden MLP)
########################################

class CNNBranch1(nn.Module):
    """
    CNN branch for (60,4).
    """
    def __init__(self, filters, kernel_size, dense_units):
        super().__init__()
        self.conv = nn.Conv1d(4, filters, kernel_size, padding=(kernel_size-1)//2)
        self.pool = nn.MaxPool1d(2)
        self.fc = nn.Linear(filters*30, dense_units)
    def forward(self, x):
        x = x.permute(0,2,1)  # (batch,4,30)
        x = F.relu(self.conv(x))
        x = self.pool(x)      # (batch,filters,30)
        x = x.flatten(start_dim=1)
        x = F.relu(self.fc(x))
        return x

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch_geometric.nn import NNConv, global_mean_pool, global_max_pool


########################################
# 4. Final Fusion Model
########################################

class CNN1OnlyModel(nn.Module):
    def __init__(self, filters1, kernel_size1, dense_units1, final_fc_dim, dropout_rate=0.0):
        super().__init__()
        self.cnn_branch1 = CNNBranch1(filters1, kernel_size1, dense_units1)
        self.dropout_rate = dropout_rate
        self.hidden = nn.Linear(dense_units1, final_fc_dim)
        self.out = nn.Linear(final_fc_dim, 1)

    def forward(self, x1):
        feat1 = self.cnn_branch1(x1)
        merged = F.dropout(feat1, p=self.dropout_rate, training=self.training)
        x = F.relu(self.hidden(merged))
        return self.out(x).view(-1)

########################################
# 5. Hybrid Dataset
########################################

class HybridDataset(Dataset):
    def __init__(self, X1, reaction_rates):
        super().__init__()
        self.X1 = X1
        self.reaction_rates = reaction_rates
        self.num_samples = len(reaction_rates)

    def __len__(self):
        return self.num_samples

    def __getitem__(self, idx):
        x1 = torch.tensor(self.X1[idx], dtype=torch.float)
        y_val = torch.tensor([self.reaction_rates[idx]], dtype=torch.float)
        return x1, y_val


def hybrid_collate(batch):
    x1_list, y_list = zip(*batch)
    x1 = torch.stack(x1_list, dim=0)
    y = torch.tensor(y_list, dtype=torch.float).view(-1)
    return x1, y

########################################
# 6. Objective Function (no pretraining)
########################################

import copy

def objective(trial):
    # Hyperparameters
    filters1 = trial.suggest_int("filters1", 16, 128, step=16)
    kernel_size1 = trial.suggest_categorical("kernel_size1", [3, 5, 7])
    dense_units1 = trial.suggest_int("dense_units1", 16, 256, step=16)
    final_fc_dim = trial.suggest_int("final_fc_dim", 16, 256, step=16)
    dropout_rate = trial.suggest_float("dropout_rate", 0.0, 0.5)
    lr = trial.suggest_loguniform("lr", 1e-4, 1e-2)
    batch_size = trial.suggest_categorical("batch_size", [8, 16, 32])
    epochs = 1000
    patience = 10
    loss_fn = nn.MSELoss()

    # Load data
    X1_HTCas9 = load_branch1_data_HTCas9()
    X1_HT11 = load_branch1_data_HT11()
    X1_mmc4 = load_branch1_data_mmc4()
    X1_mmc5 = load_branch1_data_mmc5()
    X1_change_seq = load_branch1_data_change_seq()
    X1_tiger = load_branch1_data_tiger()
    X1_iMeta = load_branch1_data_iMeta()
    
    X1_HTCas9 = np.asarray(X1_HTCas9)
    X1_HT11   = np.asarray(X1_HT11)
    X1_mmc4   = np.asarray(X1_mmc4)
    X1_mmc5   = np.asarray(X1_mmc5)
    X1_change_seq   = np.asarray(X1_change_seq)
    X1_tiger   = np.asarray(X1_tiger)
    X1_iMeta   = np.asarray(X1_iMeta)
    X1 = np.concatenate([X1_HTCas9, X1_HT11, X1_mmc4, X1_mmc5, X1_change_seq, X1_tiger, X1_iMeta], axis=0) 

    rates_HTCas9 = load_reaction_rates_HTCas9()
    rates_HT11 = load_reaction_rates_HT11()
    rates_mmc4 = load_reaction_rates_mmc4()
    rates_mmc5 = load_reaction_rates_mmc5()
    rates_change_seq = load_reaction_rates_change_seq()
    rates_tiger = load_reaction_rates_tiger()
    rates_iMeta = load_reaction_rates_iMeta()
    
    rates_HTCas9 = np.asarray(rates_HTCas9)
    rates_HT11   = np.asarray(rates_HT11)
    rates_mmc4 = np.asarray(rates_mmc4)
    rates_mmc5   = np.asarray(rates_mmc5)
    rates_change_seq   = np.asarray(rates_change_seq)
    rates_tiger   = np.asarray(rates_tiger)
    rates_iMeta   = np.asarray(rates_iMeta)
    
    rates = np.concatenate([rates_HTCas9, rates_HT11, rates_mmc4, rates_mmc5, rates_change_seq, rates_tiger, rates_iMeta], axis=0)  

    hybrid_dataset = HybridDataset(X1, rates)

    # Train/val split
    # Splitting the HTCas9 dataset
    np.random.seed(42)
    full_indices_HTCas9 = np.arange(len(rates_HTCas9))
    selected_indices_HTCas9 = np.random.choice(len(full_indices_HTCas9), size=10682, replace=False)
    unseen_indices_HTCas9 = np.setdiff1d(full_indices_HTCas9, selected_indices_HTCas9)
    train_idx_HTCas9, val_idx_HTCas9 = train_test_split(selected_indices_HTCas9, test_size=0.2, random_state=42)
    
    train_set_HTCas9 = Subset(hybrid_dataset, train_idx_HTCas9)
    val_set_HTCas9 = Subset(hybrid_dataset, val_idx_HTCas9)

    # Splitting the HT11 dataset
    np.random.seed(42)
    full_indices_HT11 = np.arange(len(rates_HT11))
    selected_indices_HT11 = np.random.choice(len(full_indices_HT11), size=9092, replace=False)
    unseen_indices_HT11 = np.setdiff1d(full_indices_HT11, selected_indices_HT11)
    # how many samples belong to HTCas9 at the front of hybrid_dataset
    offset = len(rates_HTCas9) 
    train_idx_HT11, val_idx_HT11 = train_test_split(selected_indices_HT11, test_size=0.2, random_state=42)
    # shift HT11 indices into the global index space of hybrid_dataset
    train_idx_HT11_global = (train_idx_HT11 + offset).tolist()
    val_idx_HT11_global   = (val_idx_HT11   + offset).tolist()
    # now these Subsets will point to the HT11 region of hybrid_dataset
    train_set_HT11 = Subset(hybrid_dataset, train_idx_HT11_global)
    val_set_HT11   = Subset(hybrid_dataset, val_idx_HT11_global)

    # Splitting the mmc4 dataset
    np.random.seed(42)
    full_indices_mmc4 = np.arange(len(rates_mmc4))
    selected_indices_mmc4 = np.random.choice(len(full_indices_mmc4), size=1742, replace=False)
    unseen_indices_mmc4 = np.setdiff1d(full_indices_mmc4, selected_indices_mmc4)
    # how many samples belong to HTCas9 at the front of hybrid_dataset
    offset = len(rates_HTCas9) + len(rates_HT11)
    train_idx_mmc4, val_idx_mmc4 = train_test_split(selected_indices_mmc4, test_size=0.2, random_state=42)
    # shift mmc4 indices into the global index space of hybrid_dataset
    train_idx_mmc4_global = (train_idx_mmc4 + offset).tolist()
    val_idx_mmc4_global   = (val_idx_mmc4   + offset).tolist()
    # now these Subsets will point to the mmc4 region of hybrid_dataset
    train_set_mmc4 = Subset(hybrid_dataset, train_idx_mmc4_global)
    val_set_mmc4   = Subset(hybrid_dataset, val_idx_mmc4_global)

    # Splitting the mmc5 dataset
    np.random.seed(42)
    full_indices_mmc5 = np.arange(len(rates_mmc5))
    selected_indices_mmc5 = np.random.choice(len(full_indices_mmc5), size=1835, replace=False)
    unseen_indices_mmc5 = np.setdiff1d(full_indices_mmc5, selected_indices_mmc5)
    # how many samples belong to HTCas9 at the front of hybrid_dataset
    offset = len(rates_HTCas9) + len(rates_HT11) + len(rates_mmc4)
    train_idx_mmc5, val_idx_mmc5 = train_test_split(selected_indices_mmc5, test_size=0.2, random_state=42)
    # shift mmc5 indices into the global index space of hybrid_dataset
    train_idx_mmc5_global = (train_idx_mmc5 + offset).tolist()
    val_idx_mmc5_global   = (val_idx_mmc5   + offset).tolist()
    # now these Subsets will point to the mmc5 region of hybrid_dataset
    train_set_mmc5 = Subset(hybrid_dataset, train_idx_mmc5_global)
    val_set_mmc5   = Subset(hybrid_dataset, val_idx_mmc5_global)

    # Splitting the change_seq dataset
    np.random.seed(42)
    full_indices_change_seq = np.arange(len(rates_change_seq))
    selected_indices_change_seq = np.random.choice(len(full_indices_change_seq), size=8000, replace=False)
    unseen_indices_change_seq = np.setdiff1d(full_indices_change_seq, selected_indices_change_seq)
    # how many samples belong to HTCas9 at the front of hybrid_dataset
    offset = len(rates_HTCas9) + len(rates_HT11) + len(rates_mmc4) + len(rates_mmc5)
    train_idx_change_seq, val_idx_change_seq = train_test_split(selected_indices_change_seq, test_size=0.2, random_state=42)
    # shift change_seq indices into the global index space of hybrid_dataset
    train_idx_change_seq_global = (train_idx_change_seq + offset).tolist()
    val_idx_change_seq_global   = (val_idx_change_seq   + offset).tolist()
    # now these Subsets will point to the change_seq region of hybrid_dataset
    train_set_change_seq = Subset(hybrid_dataset, train_idx_change_seq_global)
    val_set_change_seq   = Subset(hybrid_dataset, val_idx_change_seq_global)

    # Splitting the tiger dataset
    np.random.seed(42)
    full_indices_tiger = np.arange(len(rates_tiger))
    selected_indices_tiger = np.random.choice(len(full_indices_tiger), size=8000, replace=False)
    unseen_indices_tiger = np.setdiff1d(full_indices_tiger, selected_indices_tiger)
    # how many samples belong to HTCas9 at the front of hybrid_dataset
    offset = len(rates_HTCas9) + len(rates_HT11) + len(rates_mmc4) + len(rates_mmc5) + len(rates_change_seq)
    train_idx_tiger, val_idx_tiger = train_test_split(selected_indices_tiger, test_size=0.2, random_state=42)
    # shift tiger indices into the global index space of hybrid_dataset
    train_idx_tiger_global = (train_idx_tiger + offset).tolist()
    val_idx_tiger_global   = (val_idx_tiger   + offset).tolist()
    # now these Subsets will point to the tiger region of hybrid_dataset
    train_set_tiger = Subset(hybrid_dataset, train_idx_tiger_global)
    val_set_tiger   = Subset(hybrid_dataset, val_idx_tiger_global)

    # Splitting the iMeta dataset
    np.random.seed(42)
    full_indices_iMeta = np.arange(len(rates_iMeta))
    selected_indices_iMeta = np.random.choice(len(full_indices_iMeta), size=5776, replace=False)
    unseen_indices_iMeta = np.setdiff1d(full_indices_iMeta, selected_indices_iMeta)
    # how many samples belong to HTCas9 at the front of hybrid_dataset
    offset = len(rates_HTCas9) + len(rates_HT11) + len(rates_mmc4) + len(rates_mmc5) + len(rates_change_seq) + len(rates_tiger)
    train_idx_iMeta, val_idx_iMeta = train_test_split(selected_indices_iMeta, test_size=0.2, random_state=42)
    # shift iMeta indices into the global index space of hybrid_dataset
    train_idx_iMeta_global = (train_idx_iMeta + offset).tolist()
    val_idx_iMeta_global   = (val_idx_iMeta   + offset).tolist()
    # now these Subsets will point to the iMeta region of hybrid_dataset
    train_set_iMeta = Subset(hybrid_dataset, train_idx_iMeta_global)
    val_set_iMeta   = Subset(hybrid_dataset, val_idx_iMeta_global)

    from torch.utils.data import ConcatDataset, DataLoader

    # combine the subsets (datasets)
    train_set = ConcatDataset([train_set_HTCas9, train_set_HT11, train_set_mmc4, train_set_mmc5, train_set_change_seq, train_set_tiger, train_set_iMeta])
    val_set   = ConcatDataset([val_set_HTCas9,   val_set_HT11,   val_set_mmc4,   val_set_mmc5, val_set_change_seq, val_set_tiger, val_set_iMeta])
    
    # build new loaders
    train_loader = DataLoader(train_set, batch_size=batch_size, shuffle=True, collate_fn=hybrid_collate)
    val_loader   = DataLoader(val_set,   batch_size=batch_size, shuffle=False, collate_fn=hybrid_collate)

    # Build model
    model = CNN1OnlyModel(filters1, kernel_size1, dense_units1, final_fc_dim, dropout_rate)
    optimizer = optim.Adam(model.parameters(), lr=lr, weight_decay=1e-4)

    best_val_loss = float('inf')
    counter = 0
    best_model_state = None

    for epoch in range(epochs):
        model.train()
        total_loss = 0.0
        for x1_b, y_b in train_loader:
            pred = model(x1_b)
            loss = loss_fn(pred, y_b)
            optimizer.zero_grad()
            loss.backward()
            optimizer.step()
            total_loss += loss.item()
        avg_train_loss = total_loss / len(train_loader)

        model.eval()
        total_val = 0.0
        with torch.no_grad():
            for x1_b, y_b in val_loader:
                valpred = model(x1_b)
                vloss = loss_fn(valpred, y_b)
                total_val += vloss.item()
        avg_val_loss = total_val / len(val_loader)
        trial.report(avg_val_loss, epoch)

        if avg_val_loss < best_val_loss:
            best_val_loss = avg_val_loss
            best_model_state = copy.deepcopy(model)
            counter = 0
        else:
            counter += 1
            
        if counter >= patience:
            print(f"Early stopping at epoch {epoch+1}")
            break

        print(f"Epoch {epoch+1}/{epochs} | Train: {avg_train_loss:.4f} | Val: {avg_val_loss:.4f}")

    if best_model_state is not None:
        model = best_model_state
    torch.save(model, f"trial_{trial.number}_model_HTCas9_HT11_RfxCas13d_change_seq_sampled_2_tiger_iMeta_combined_unique_CNN1_only_reduced_feature.pt")

    # how many samples at the front of hybrid_dataset
    offset_1 = len(rates_HTCas9)
    offset_2 = len(rates_HTCas9) + len(rates_HT11)
    offset_3 = len(rates_HTCas9) + len(rates_HT11) + len(rates_mmc4)
    offset_4 = len(rates_HTCas9) + len(rates_HT11) + len(rates_mmc4) + len(rates_mmc5)
    offset_5 = len(rates_HTCas9) + len(rates_HT11) + len(rates_mmc4) + len(rates_mmc5) + len(rates_change_seq)
    offset_6 = len(rates_HTCas9) + len(rates_HT11) + len(rates_mmc4) + len(rates_mmc5) + len(rates_change_seq) + len(rates_tiger)
    # shift indices into the global index space of hybrid_dataset
    unseen_indices_HT11_global = (unseen_indices_HT11 + offset_1).tolist()
    unseen_indices_mmc4_global = (unseen_indices_mmc4 + offset_2).tolist()
    unseen_indices_mmc5_global = (unseen_indices_mmc5 + offset_3).tolist()
    unseen_indices_change_seq_global = (unseen_indices_change_seq + offset_4).tolist()
    unseen_indices_tiger_global = (unseen_indices_tiger + offset_5).tolist()
    unseen_indices_iMeta_global = (unseen_indices_iMeta + offset_6).tolist()
    
    # Evaluate on 10 subsamples from unseen
    unseen_set_HTCas9 = Subset(hybrid_dataset, unseen_indices_HTCas9)
    unseen_set_HT11 = Subset(hybrid_dataset, unseen_indices_HT11_global)
    unseen_set_mmc4 = Subset(hybrid_dataset, unseen_indices_mmc4_global)
    unseen_set_mmc5 = Subset(hybrid_dataset, unseen_indices_mmc5_global)
    unseen_set_change_seq = Subset(hybrid_dataset, unseen_indices_change_seq_global)
    unseen_set_tiger = Subset(hybrid_dataset, unseen_indices_tiger_global)
    unseen_set_iMeta = Subset(hybrid_dataset, unseen_indices_iMeta_global)
        
    model.eval()

    # Compute unseen Spearman for HTCas9 dataset
    spearman_list_HTCas9 = []
    for i in range(10):
        if len(unseen_set_HTCas9) < 1000:
            break
        idx_trial = np.random.choice(len(unseen_set_HTCas9), size=1000, replace=False)
        trial_subset = Subset(unseen_set_HTCas9, idx_trial)
        trial_loader = DataLoader(trial_subset, batch_size=1, shuffle=False, collate_fn=hybrid_collate)
        
        preds_trial = []
        labels_trial = []
        with torch.no_grad():
            for xx1_b, yy_b in trial_loader:
                p = model(xx1_b)
                preds_trial.append(p.item())
                labels_trial.append(yy_b.item())
        sp_corr, _ = spearmanr(labels_trial, preds_trial)
        spearman_list_HTCas9.append(sp_corr)
        print(f"Subsample {i+1}: Spearman = {sp_corr:.4f}")
    avg_spearman = np.mean(spearman_list_HTCas9) if spearman_list_HTCas9 else None
    print(f"Average Spearman over 10 subsamples in HTCas9 dataset: {avg_spearman}")


    # Compute unseen Spearman for HT11 dataset
    spearman_list_HT11 = []
    for i in range(10):
        if len(unseen_set_HT11) < 1000:
            break
        idx_trial = np.random.choice(len(unseen_set_HT11), size=1000, replace=False)
        trial_subset = Subset(unseen_set_HT11, idx_trial)
        trial_loader = DataLoader(trial_subset, batch_size=1, shuffle=False, collate_fn=hybrid_collate)
        
        preds_trial = []
        labels_trial = []
        with torch.no_grad():
            for xx1_b, yy_b in trial_loader:
                p = model(xx1_b)
                preds_trial.append(p.item())
                labels_trial.append(yy_b.item())
        sp_corr, _ = spearmanr(labels_trial, preds_trial)
        spearman_list_HT11.append(sp_corr)
        print(f"Subsample {i+1}: Spearman = {sp_corr:.4f}")
    avg_spearman = np.mean(spearman_list_HT11) if spearman_list_HT11 else None
    print(f"Average Spearman over 10 subsamples in HT11 dataset: {avg_spearman}")

    # Compute unseen Spearman for mmc4 dataset
    spearman_list_mmc4 = []
    for i in range(10):
        if len(unseen_set_mmc4) < 300:
            break
        idx_trial = np.random.choice(len(unseen_set_mmc4), size=300, replace=False)
        trial_subset = Subset(unseen_set_mmc4, idx_trial)
        trial_loader = DataLoader(trial_subset, batch_size=1, shuffle=False, collate_fn=hybrid_collate)
        
        preds_trial = []
        labels_trial = []
        with torch.no_grad():
            for xx1_b, yy_b in trial_loader:
                p = model(xx1_b)
                preds_trial.append(p.item())
                labels_trial.append(yy_b.item())
        sp_corr, _ = spearmanr(labels_trial, preds_trial)
        spearman_list_mmc4.append(sp_corr)
        print(f"Subsample {i+1}: Spearman = {sp_corr:.4f}")
    avg_spearman = np.mean(spearman_list_mmc4) if spearman_list_mmc4 else None
    print(f"Average Spearman over 10 subsamples in mmc4 dataset: {avg_spearman}")

    # Compute unseen Spearman for mmc5 dataset
    spearman_list_mmc5 = []
    for i in range(10):
        if len(unseen_set_mmc5) < 300:
            break
        idx_trial = np.random.choice(len(unseen_set_mmc5), size=300, replace=False)
        trial_subset = Subset(unseen_set_mmc5, idx_trial)
        trial_loader = DataLoader(trial_subset, batch_size=1, shuffle=False, collate_fn=hybrid_collate)
        
        preds_trial = []
        labels_trial = []
        with torch.no_grad():
            for xx1_b, yy_b in trial_loader:
                p = model(xx1_b)
                preds_trial.append(p.item())
                labels_trial.append(yy_b.item())
        sp_corr, _ = spearmanr(labels_trial, preds_trial)
        spearman_list_mmc5.append(sp_corr)
        print(f"Subsample {i+1}: Spearman = {sp_corr:.4f}")
    avg_spearman = np.mean(spearman_list_mmc5) if spearman_list_mmc5 else None
    print(f"Average Spearman over 10 subsamples in mmc5 dataset: {avg_spearman}")

    # Compute unseen Spearman for change_seq dataset
    spearman_list_change_seq = []
    for i in range(10):
        if len(unseen_set_change_seq) < 1000:
            break
        idx_trial = np.random.choice(len(unseen_set_change_seq), size=1000, replace=False)
        trial_subset = Subset(unseen_set_change_seq, idx_trial)
        trial_loader = DataLoader(trial_subset, batch_size=1, shuffle=False, collate_fn=hybrid_collate)
        
        preds_trial = []
        labels_trial = []
        with torch.no_grad():
            for xx1_b, yy_b in trial_loader:
                p = model(xx1_b)
                preds_trial.append(p.item())
                labels_trial.append(yy_b.item())
        sp_corr, _ = spearmanr(labels_trial, preds_trial)
        spearman_list_change_seq.append(sp_corr)
        print(f"Subsample {i+1}: Spearman = {sp_corr:.4f}")
    avg_spearman = np.mean(spearman_list_change_seq) if spearman_list_change_seq else None
    print(f"Average Spearman over 10 subsamples in change_seq dataset: {avg_spearman}")

    # Compute unseen Spearman for tiger dataset
    spearman_list_tiger = []
    for i in range(10):
        if len(unseen_set_tiger) < 1000:
            break
        idx_trial = np.random.choice(len(unseen_set_tiger), size=1000, replace=False)
        trial_subset = Subset(unseen_set_tiger, idx_trial)
        trial_loader = DataLoader(trial_subset, batch_size=1, shuffle=False, collate_fn=hybrid_collate)
        
        preds_trial = []
        labels_trial = []
        with torch.no_grad():
            for xx1_b, yy_b in trial_loader:
                p = model(xx1_b)
                preds_trial.append(p.item())
                labels_trial.append(yy_b.item())
        sp_corr, _ = spearmanr(labels_trial, preds_trial)
        spearman_list_tiger.append(sp_corr)
        print(f"Subsample {i+1}: Spearman = {sp_corr:.4f}")
    avg_spearman = np.mean(spearman_list_tiger) if spearman_list_tiger else None   
    print(f"Average Spearman over 10 subsamples in tiger dataset: {avg_spearman}")

    # Compute unseen Spearman for iMeta dataset
    spearman_list_iMeta = []
    for i in range(10):
        if len(unseen_set_iMeta) < 1000:
            break
        idx_trial = np.random.choice(len(unseen_set_iMeta), size=1000, replace=False)
        trial_subset = Subset(unseen_set_iMeta, idx_trial)
        trial_loader = DataLoader(trial_subset, batch_size=1, shuffle=False, collate_fn=hybrid_collate)
        
        preds_trial = []
        labels_trial = []
        with torch.no_grad():
            for xx1_b, yy_b in trial_loader:
                p = model(xx1_b)

                preds_trial.append(p.item())
                labels_trial.append(yy_b.item())
        sp_corr, _ = spearmanr(labels_trial, preds_trial)
        spearman_list_iMeta.append(sp_corr)
        print(f"Subsample {i+1}: Spearman = {sp_corr:.4f}")
    avg_spearman = np.mean(spearman_list_iMeta) if spearman_list_iMeta else None   
    print(f"Average Spearman over 10 subsamples in iMeta dataset: {avg_spearman}")
    
    return best_val_loss


########################################
# 7. Main Pipeline
########################################

def main_pipeline():
    study = optuna.create_study(direction="minimize")
    study.optimize(objective, n_trials=10)
    
    print("Best hyperparameters:", study.best_trial.params)
    best_trial_number = study.best_trial.number
    best_params = study.best_trial.params
    
    # Rebuild for final 100-trial evaluation by loading the entire saved model
    model = torch.load(f"trial_{best_trial_number}_model_HTCas9_HT11_RfxCas13d_change_seq_sampled_2_tiger_iMeta_combined_unique_CNN1_only_reduced_feature.pt", weights_only=False)
    model.eval()
    
    # final hold-out evaluation
    # Load data
    X1_HTCas9 = load_branch1_data_HTCas9()
    X1_HT11 = load_branch1_data_HT11()
    X1_mmc4 = load_branch1_data_mmc4()
    X1_mmc5 = load_branch1_data_mmc5()
    X1_change_seq = load_branch1_data_change_seq()
    X1_tiger = load_branch1_data_tiger()
    X1_iMeta = load_branch1_data_iMeta()
    
    X1_HTCas9 = np.asarray(X1_HTCas9)
    X1_HT11   = np.asarray(X1_HT11)
    X1_mmc4 = np.asarray(X1_mmc4)
    X1_mmc5   = np.asarray(X1_mmc5)
    X1_change_seq   = np.asarray(X1_change_seq)
    X1_tiger   = np.asarray(X1_tiger)
    X1_iMeta   = np.asarray(X1_iMeta)
    
    X1 = np.concatenate([X1_HTCas9, X1_HT11, X1_mmc4, X1_mmc5, X1_change_seq, X1_tiger, X1_iMeta], axis=0) 

    rates_HTCas9 = load_reaction_rates_HTCas9()
    rates_HT11 = load_reaction_rates_HT11()
    rates_mmc4 = load_reaction_rates_mmc4()
    rates_mmc5 = load_reaction_rates_mmc5()
    rates_change_seq = load_reaction_rates_change_seq()
    rates_tiger = load_reaction_rates_tiger()
    rates_iMeta = load_reaction_rates_iMeta()
    
    rates_HTCas9 = np.asarray(rates_HTCas9)
    rates_HT11   = np.asarray(rates_HT11)
    rates_mmc4 = np.asarray(rates_mmc4)
    rates_mmc5   = np.asarray(rates_mmc5)
    rates_change_seq   = np.asarray(rates_change_seq)
    rates_tiger   = np.asarray(rates_tiger)
    rates_iMeta   = np.asarray(rates_iMeta)
    
    rates = np.concatenate([rates_HTCas9, rates_HT11, rates_mmc4, rates_mmc5, rates_change_seq, rates_tiger, rates_iMeta], axis=0)
    
    hybrid_dataset = HybridDataset(X1, rates)
    
    # Get unseen HTCas9 dataset
    np.random.seed(42)
    full_indices_HTCas9 = np.arange(len(rates_HTCas9))
    selected_indices_HTCas9 = np.random.choice(len(full_indices_HTCas9), size=10682, replace=False)
    unseen_indices_HTCas9 = np.setdiff1d(full_indices_HTCas9, selected_indices_HTCas9)

    # Splitting the HT11 dataset
    np.random.seed(42)
    full_indices_HT11 = np.arange(len(rates_HT11))
    selected_indices_HT11 = np.random.choice(len(full_indices_HT11), size=9092, replace=False)
    unseen_indices_HT11 = np.setdiff1d(full_indices_HT11, selected_indices_HT11)
    # how many samples belong to HTCas9 at the front of hybrid_dataset
    offset = len(rates_HTCas9)
    # shift HT11 indices into the global index space of hybrid_dataset
    unseen_indices_HT11_global = (unseen_indices_HT11 + offset).tolist()

    # Splitting the mmc4 dataset
    np.random.seed(42)
    full_indices_mmc4 = np.arange(len(rates_mmc4))
    selected_indices_mmc4 = np.random.choice(len(full_indices_mmc4), size=1742, replace=False)
    unseen_indices_mmc4 = np.setdiff1d(full_indices_mmc4, selected_indices_mmc4)
    # how many samples belong to HTCas9 at the front of hybrid_dataset
    offset = len(rates_HTCas9) + len(rates_HT11)
    # shift mmc4 indices into the global index space of hybrid_dataset
    unseen_indices_mmc4_global = (unseen_indices_mmc4 + offset).tolist()

    # Splitting the mmc5 dataset
    np.random.seed(42)
    full_indices_mmc5 = np.arange(len(rates_mmc5))
    selected_indices_mmc5 = np.random.choice(len(full_indices_mmc5), size=1835, replace=False)
    unseen_indices_mmc5 = np.setdiff1d(full_indices_mmc5, selected_indices_mmc5)
    # how many samples belong to HTCas9 at the front of hybrid_dataset
    offset = len(rates_HTCas9) + len(rates_HT11) + len(rates_mmc4)
    # shift mmc5 indices into the global index space of hybrid_dataset
    unseen_indices_mmc5_global = (unseen_indices_mmc5 + offset).tolist()

    # Splitting the change_seq dataset
    np.random.seed(42)
    full_indices_change_seq = np.arange(len(rates_change_seq))
    selected_indices_change_seq = np.random.choice(len(full_indices_change_seq), size=8000, replace=False)
    unseen_indices_change_seq = np.setdiff1d(full_indices_change_seq, selected_indices_change_seq)
    # how many samples belong to HTCas9 at the front of hybrid_dataset
    offset = len(rates_HTCas9) + len(rates_HT11) + len(rates_mmc4) + len(rates_mmc5)
    # shift change_seq indices into the global index space of hybrid_dataset
    unseen_indices_change_seq_global = (unseen_indices_change_seq + offset).tolist()

    # Splitting the tiger dataset
    np.random.seed(42)
    full_indices_tiger = np.arange(len(rates_tiger))
    selected_indices_tiger = np.random.choice(len(full_indices_tiger), size=8000, replace=False)
    unseen_indices_tiger = np.setdiff1d(full_indices_tiger, selected_indices_tiger)
    # how many samples belong to HTCas9 at the front of hybrid_dataset
    offset = len(rates_HTCas9) + len(rates_HT11) + len(rates_mmc4) + len(rates_mmc5) + len(rates_change_seq)
    # shift tiger indices into the global index space of hybrid_dataset
    unseen_indices_tiger_global = (unseen_indices_tiger + offset).tolist()

    # Splitting the iMeta dataset
    np.random.seed(42)
    full_indices_iMeta = np.arange(len(rates_iMeta))
    selected_indices_iMeta = np.random.choice(len(full_indices_iMeta), size=5776, replace=False)
    unseen_indices_iMeta = np.setdiff1d(full_indices_iMeta, selected_indices_iMeta)
    # how many samples belong to HTCas9 at the front of hybrid_dataset
    offset = len(rates_HTCas9) + len(rates_HT11) + len(rates_mmc4) + len(rates_mmc5) + len(rates_change_seq) + len(rates_tiger)
    # shift iMeta indices into the global index space of hybrid_dataset
    unseen_indices_iMeta_global = (unseen_indices_iMeta + offset).tolist()

    from torch.utils.data import ConcatDataset, DataLoader

    # Evaluate on 10 subsamples from unseen
    unseen_set_HTCas9 = Subset(hybrid_dataset, unseen_indices_HTCas9)
    unseen_set_HT11 = Subset(hybrid_dataset, unseen_indices_HT11_global)
    unseen_set_mmc4 = Subset(hybrid_dataset, unseen_indices_mmc4_global)
    unseen_set_mmc5 = Subset(hybrid_dataset, unseen_indices_mmc5_global)
    unseen_set_change_seq = Subset(hybrid_dataset, unseen_indices_change_seq_global)
    unseen_set_tiger = Subset(hybrid_dataset, unseen_indices_tiger_global)
    unseen_set_iMeta = Subset(hybrid_dataset, unseen_indices_iMeta_global)
        
    # Compute unseen Spearman for HTCas9 dataset
    spearman_list_HTCas9 = []
    for i in range(100):
        if len(unseen_set_HTCas9) < 1000:
            print("Not enough unseen samples to draw 1000 each time.")
            break
        idx_trial = np.random.choice(len(unseen_set_HTCas9), size=1000, replace=False)
        trial_subset = Subset(unseen_set_HTCas9, idx_trial)
        trial_loader = DataLoader(trial_subset, batch_size=1, shuffle=False, collate_fn=hybrid_collate)
        
        preds_trial = []
        labels_trial = []
        with torch.no_grad():
            for xx1_b, yy_b in trial_loader:
                p = model(xx1_b)
                preds_trial.append(p.item())
                labels_trial.append(yy_b.item())
        sp_corr, _ = spearmanr(labels_trial, preds_trial)
        spearman_list_HTCas9.append(sp_corr)
        print(sp_corr)
    avg_spearman = np.mean(spearman_list_HTCas9) if spearman_list_HTCas9 else None
    print(f"Average Spearman Correlation over 100 trials in HTCas9 dataset: {avg_spearman}")


    # Compute unseen Spearman for HT11 dataset
    spearman_list_HT11 = []
    for i in range(100):
        if len(unseen_set_HT11) < 1000:
            print("Not enough unseen samples to draw 1000 each time.")
            break
        idx_trial = np.random.choice(len(unseen_set_HT11), size=1000, replace=False)
        trial_subset = Subset(unseen_set_HT11, idx_trial)
        trial_loader = DataLoader(trial_subset, batch_size=1, shuffle=False, collate_fn=hybrid_collate)
        
        preds_trial = []
        labels_trial = []
        with torch.no_grad():
            for xx1_b, yy_b in trial_loader:
                p = model(xx1_b)
                preds_trial.append(p.item())
                labels_trial.append(yy_b.item())
        sp_corr, _ = spearmanr(labels_trial, preds_trial)
        spearman_list_HT11.append(sp_corr)
        print(sp_corr)
    avg_spearman = np.mean(spearman_list_HT11) if spearman_list_HT11 else None
    print(f"Average Spearman Correlation over 100 trials in HT11 dataset: {avg_spearman}")

    # Compute unseen Spearman for mmc4 dataset
    spearman_list_mmc4 = []
    for i in range(100):
        if len(unseen_set_mmc4) < 300:
            print("Not enough unseen samples to draw 300 each time.")
            break
        idx_trial = np.random.choice(len(unseen_set_mmc4), size=300, replace=False)
        trial_subset = Subset(unseen_set_mmc4, idx_trial)
        trial_loader = DataLoader(trial_subset, batch_size=1, shuffle=False, collate_fn=hybrid_collate)
        
        preds_trial = []
        labels_trial = []
        with torch.no_grad():
            for xx1_b, yy_b in trial_loader:
                p = model(xx1_b)
                preds_trial.append(p.item())
                labels_trial.append(yy_b.item())
        sp_corr, _ = spearmanr(labels_trial, preds_trial)
        spearman_list_mmc4.append(sp_corr)
        print(sp_corr)
    avg_spearman = np.mean(spearman_list_mmc4) if spearman_list_mmc4 else None
    print(f"Average Spearman Correlation over 100 trials in mmc4 dataset: {avg_spearman}")

    # Compute unseen Spearman for mmc5 dataset
    spearman_list_mmc5 = []
    for i in range(100):
        if len(unseen_set_mmc5) < 300:
            print("Not enough unseen samples to draw 300 each time.")
            break
        idx_trial = np.random.choice(len(unseen_set_mmc5), size=300, replace=False)
        trial_subset = Subset(unseen_set_mmc5, idx_trial)
        trial_loader = DataLoader(trial_subset, batch_size=1, shuffle=False, collate_fn=hybrid_collate)
        
        preds_trial = []
        labels_trial = []
        with torch.no_grad():
            for xx1_b, yy_b in trial_loader:
                p = model(xx1_b)
                preds_trial.append(p.item())
                labels_trial.append(yy_b.item())
        sp_corr, _ = spearmanr(labels_trial, preds_trial)
        spearman_list_mmc5.append(sp_corr)
        print(sp_corr)
    avg_spearman = np.mean(spearman_list_mmc5) if spearman_list_mmc5 else None
    print(f"Average Spearman Correlation over 100 trials in mmc5 dataset: {avg_spearman}")

    # Compute unseen Spearman for change_seq dataset
    spearman_list_change_seq = []
    for i in range(100):
        if len(unseen_set_change_seq) < 1000:
            print("Not enough unseen samples to draw 1000 each time.")
            break
        idx_trial = np.random.choice(len(unseen_set_change_seq), size=1000, replace=False)
        trial_subset = Subset(unseen_set_change_seq, idx_trial)
        trial_loader = DataLoader(trial_subset, batch_size=1, shuffle=False, collate_fn=hybrid_collate)
        preds_trial = []
        labels_trial = []
        with torch.no_grad():
            for xx1_b, yy_b in trial_loader:
                p = model(xx1_b)
                preds_trial.append(p.item())
                labels_trial.append(yy_b.item())
        sp_corr, _ = spearmanr(labels_trial, preds_trial)
        spearman_list_change_seq.append(sp_corr)
        print(sp_corr)

    avg_spearman = np.mean(spearman_list_change_seq) if spearman_list_change_seq else None
    print(f"Average Spearman Correlation over 100 trials in change_seq dataset: {avg_spearman}")

    # Compute unseen Spearman for tiger dataset
    spearman_list_tiger = []
    for i in range(100):
        if len(unseen_set_tiger) < 1000:
            print("Not enough unseen samples to draw 1000 each time.")
            break
        idx_trial = np.random.choice(len(unseen_set_tiger), size=1000, replace=False)
        trial_subset = Subset(unseen_set_tiger, idx_trial)
        trial_loader = DataLoader(trial_subset, batch_size=1, shuffle=False, collate_fn=hybrid_collate)
        preds_trial = []
        labels_trial = []
        with torch.no_grad():
            for xx1_b, yy_b in trial_loader:
                p = model(xx1_b)
                preds_trial.append(p.item())
                labels_trial.append(yy_b.item())
        sp_corr, _ = spearmanr(labels_trial, preds_trial)
        spearman_list_tiger.append(sp_corr)
        print(sp_corr)

    avg_spearman = np.mean(spearman_list_tiger) if spearman_list_tiger else None
    print(f"Average Spearman Correlation over 100 trials in tiger dataset: {avg_spearman}")

    # Compute unseen Spearman for iMeta dataset
    spearman_list_iMeta = []
    for i in range(100):
        if len(unseen_set_iMeta) < 1000:
            print("Not enough unseen samples to draw 1000 each time.")

            break
        idx_trial = np.random.choice(len(unseen_set_iMeta), size=1000, replace=False)
        trial_subset = Subset(unseen_set_iMeta, idx_trial)
        trial_loader = DataLoader(trial_subset, batch_size=1, shuffle=False, collate_fn=hybrid_collate)
        preds_trial = []
        labels_trial = []
        with torch.no_grad():
            for xx1_b, yy_b in trial_loader:
                p = model(xx1_b)
                preds_trial.append(p.item())
                labels_trial.append(yy_b.item())
        sp_corr, _ = spearmanr(labels_trial, preds_trial)
        spearman_list_iMeta.append(sp_corr)
        print(sp_corr)

    avg_spearman = np.mean(spearman_list_iMeta) if spearman_list_iMeta else None
    print(f"Average Spearman Correlation over 100 trials in iMeta dataset: {avg_spearman}")

if __name__ == "__main__":
    main_pipeline()
```
