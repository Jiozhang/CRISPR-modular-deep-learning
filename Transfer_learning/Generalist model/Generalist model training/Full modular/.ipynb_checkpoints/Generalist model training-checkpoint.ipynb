```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
import numpy as np

from torch.utils.data import Dataset, DataLoader, Subset
from sklearn.model_selection import train_test_split
from scipy.stats import spearmanr
import optuna

import copy

# PyG
from torch_geometric.data import Data, Batch
from torch_geometric.nn import GCNConv, global_mean_pool

# NUPACK
from nupack import Model, pairs

np.set_printoptions(threshold=np.inf)
print("Using PyTorch and PyG.")

########################################
# 1. Data Loading
########################################

def load_branch1_data_HTCas9():
    """
    Loads (60,4) data from 'Feature_CNN1_reduced_HTCas9.txt'.
    """
    data_branch1 = []
    current_array = []
    with open('Feature_CNN1_reduced_HTCas9.txt', 'r') as f:
        for line in f:             
            line = line.strip()
            if line == "":
                if current_array:
                    data_branch1.append(current_array)
                    current_array = []
            else:
                line = line.replace('[','').replace(']', '')
                row = [float(x.strip().replace(',', '')) for x in line.split()]
                current_array.append(row)
    if current_array:
        data_branch1.append(current_array)
    X_branch1 = np.array(data_branch1)
    X_branch1 = X_branch1.reshape(len(X_branch1), 60, 4)
    return X_branch1


def load_branch1_data_HT11():
    """
    Loads (60,4) data from 'Feature_CNN1_reduced_HT11.txt'.
    """
    data_branch1 = []
    current_array = []
    with open('Feature_CNN1_reduced_HT11.txt', 'r') as f:
        for line in f:
            line = line.strip()
            if line == "":
                if current_array:
                    data_branch1.append(current_array)
                    current_array = []
            else:
                line = line.replace('[','').replace(']', '')
                row = [float(x.strip().replace(',', '')) for x in line.split()]
                current_array.append(row)
    if current_array:
        data_branch1.append(current_array)
    X_branch1 = np.array(data_branch1)
    X_branch1 = X_branch1.reshape(len(X_branch1), 60, 4)
    return X_branch1

def load_branch1_data_mmc4():

    """
    Loads (60,4) data from 'Feature_CNN1_reduced_mmc4.txt'.
    """
    data_branch1 = []
    current_array = []
    with open('Feature_CNN1_reduced_mmc4.txt', 'r') as f:
        for line in f:
            line = line.strip()
            if line == "":
                if current_array:
                    data_branch1.append(current_array)
                    current_array = []
            else:
                line = line.replace('[','').replace(']', '')
                row = [float(x.strip().replace(',', '')) for x in line.split()]
                current_array.append(row)
    if current_array:
        data_branch1.append(current_array)
    X_branch1 = np.array(data_branch1)
    X_branch1 = X_branch1.reshape(len(X_branch1), 60, 4)
    return X_branch1


def load_branch1_data_mmc5():
    """
    Loads (60,4) data from 'Feature_CNN1_reduced_mmc5.txt'.
    """
    data_branch1 = []
    current_array = []
    with open('Feature_CNN1_reduced_mmc5.txt', 'r') as f:
        for line in f:
            line = line.strip()
            if line == "":
                if current_array:
                    data_branch1.append(current_array)
                    current_array = []
            else:
                line = line.replace('[','').replace(']', '')
                row = [float(x.strip().replace(',', '')) for x in line.split()]
                current_array.append(row)
    if current_array:
        data_branch1.append(current_array)
    X_branch1 = np.array(data_branch1)
    X_branch1 = X_branch1.reshape(len(X_branch1), 60, 4)
    return X_branch1

def load_branch1_data_change_seq():
    """
    Loads (60,4) data from 'Feature_CNN1_reduced_change_seq.txt'.
    """
    data_branch1 = []
    current_array = []
    with open('Feature_CNN1_reduced_change_seq.txt', 'r') as f:
        for line in f:
            line = line.strip()
            if line == "":
                if current_array:
                    data_branch1.append(current_array)
                    current_array = []
            else:
                line = line.replace('[','').replace(']', '')
                row = [float(x.strip().replace(',', '')) for x in line.split()]
                current_array.append(row)
    if current_array:
        data_branch1.append(current_array)
    X_branch1 = np.array(data_branch1)
    X_branch1 = X_branch1.reshape(len(X_branch1), 60, 4)
    return X_branch1

def load_branch1_data_tiger():
    """
    Loads (60,4) data from 'Feature_CNN1_reduced_tiger.txt'.
    """
    data_branch1 = []
    current_array = []
    with open('Feature_CNN1_reduced_tiger.txt', 'r') as f:
        for line in f:
            line = line.strip()
            if line == "":
                if current_array:
                    data_branch1.append(current_array)
                    current_array = []
            else:
                line = line.replace('[','').replace(']', '')
                row = [float(x.strip().replace(',', '')) for x in line.split()]
                current_array.append(row)
    if current_array:
        data_branch1.append(current_array)
    X_branch1 = np.array(data_branch1)
    X_branch1 = X_branch1.reshape(len(X_branch1), 60, 4)
    return X_branch1

def load_branch1_data_iMeta():
    """
    Loads (60,4) data from 'Feature_CNN1_reduced_iMeta.txt'.
    """
    data_branch1 = []
    current_array = []
    with open('Feature_CNN1_reduced_iMeta.txt', 'r') as f:
        for line in f:
            line = line.strip()
            if line == "":
                if current_array:
                    data_branch1.append(current_array)
                    current_array = []
            else:
                line = line.replace('[','').replace(']', '')
                row = [float(x.strip().replace(',', '')) for x in line.split()]
                current_array.append(row)
    if current_array:
        data_branch1.append(current_array)
    X_branch1 = np.array(data_branch1)
    X_branch1 = X_branch1.reshape(len(X_branch1), 60, 4)
    return X_branch1



def load_branch2_data_HTCas9():
    """
    Loads (90,3) data from 'Feature_CNN2_reduced_HTCas9.txt'.
    """
    data_branch2 = []
    current_array = []
    with open('Feature_CNN2_reduced_HTCas9.txt', 'r') as f:
        for line in f:
            line = line.strip()
            if line == "":
                if current_array:
                    data_branch2.append(current_array)
                    current_array = []
            else:
                line = line.replace('[','').replace(']', '')
                row = [float(x.strip().replace(',', '')) for x in line.split()]
                current_array.append(row)
    if current_array:
        data_branch2.append(current_array)
    X_branch2 = np.array(data_branch2)
    X_branch2 = X_branch2.reshape(len(X_branch2), 90, 3)
    return X_branch2

def load_branch2_data_HT11():
    """
    Loads (90,3) data from 'Feature_CNN2_reduced_HT11.txt'.
    """
    data_branch2 = []
    current_array = []
    with open('Feature_CNN2_reduced_HT11.txt', 'r') as f:
        for line in f:
            line = line.strip()
            if line == "":
                if current_array:
                    data_branch2.append(current_array)
                    current_array = []
            else:
                line = line.replace('[','').replace(']', '')
                row = [float(x.strip().replace(',', '')) for x in line.split()]
                current_array.append(row)
    if current_array:
        data_branch2.append(current_array)
    X_branch2 = np.array(data_branch2)
    X_branch2 = X_branch2.reshape(len(X_branch2), 90, 3)
    return X_branch2

def load_branch2_data_mmc4():
    """
    Loads (90,3) data from 'Feature_CNN2_reduced_mmc4.txt'.
    """
    data_branch2 = []
    current_array = []
    with open('Feature_CNN2_reduced_mmc4.txt', 'r') as f:
        for line in f:
            line = line.strip()
            if line == "":
                if current_array:
                    data_branch2.append(current_array)
                    current_array = []
            else:
                line = line.replace('[','').replace(']', '')
                row = [float(x.strip().replace(',', '')) for x in line.split()]
                current_array.append(row)
    if current_array:
        data_branch2.append(current_array)
    X_branch2 = np.array(data_branch2)
    X_branch2 = X_branch2.reshape(len(X_branch2), 90, 3)
    return X_branch2

def load_branch2_data_mmc5():
    """
    Loads (90,3) data from 'Feature_CNN2_reduced_mmc5.txt'.
    """
    data_branch2 = []
    current_array = []
    with open('Feature_CNN2_reduced_mmc5.txt', 'r') as f:
        for line in f:
            line = line.strip()
            if line == "":
                if current_array:
                    data_branch2.append(current_array)
                    current_array = []
            else:
                line = line.replace('[','').replace(']', '')
                row = [float(x.strip().replace(',', '')) for x in line.split()]
                current_array.append(row)
    if current_array:
        data_branch2.append(current_array)
    X_branch2 = np.array(data_branch2)
    X_branch2 = X_branch2.reshape(len(X_branch2), 90, 3)
    return X_branch2

def load_branch2_data_change_seq():
    """
    Loads (90,3) data from 'Feature_CNN2_reduced_change_seq.txt'.
    """
    data_branch2 = []
    current_array = []
    with open('Feature_CNN2_reduced_change_seq.txt', 'r') as f:
        for line in f:
            line = line.strip()
            if line == "":
                if current_array:
                    data_branch2.append(current_array)
                    current_array = []
            else:
                line = line.replace('[','').replace(']', '')
                row = [float(x.strip().replace(',', '')) for x in line.split()]
                current_array.append(row)
    if current_array:
        data_branch2.append(current_array)
    X_branch2 = np.array(data_branch2)
    X_branch2 = X_branch2.reshape(len(X_branch2), 90, 3)
    return X_branch2

def load_branch2_data_tiger():
    """
    Loads (90,3) data from 'Feature_CNN2_reduced_tiger.txt'.
    """
    data_branch2 = []
    current_array = []
    with open('Feature_CNN2_reduced_tiger.txt', 'r') as f:
        for line in f:
            line = line.strip()
            if line == "":
                if current_array:
                    data_branch2.append(current_array)
                    current_array = []
            else:
                line = line.replace('[','').replace(']', '')
                row = [float(x.strip().replace(',', '')) for x in line.split()]
                current_array.append(row)
    if current_array:
        data_branch2.append(current_array)
    X_branch2 = np.array(data_branch2)
    X_branch2 = X_branch2.reshape(len(X_branch2), 90, 3)
    return X_branch2

def load_branch2_data_iMeta():
    """
    Loads (90,3) data from 'Feature_CNN2_reduced_iMeta.txt'.
    """
    data_branch2 = []
    current_array = []
    with open('Feature_CNN2_reduced_iMeta.txt', 'r') as f:
        for line in f:
            line = line.strip()
            if line == "":
                if current_array:
                    data_branch2.append(current_array)
                    current_array = []
            else:
                line = line.replace('[','').replace(']', '')
                row = [float(x.strip().replace(',', '')) for x in line.split()]
                current_array.append(row)
    if current_array:
        data_branch2.append(current_array)
    X_branch2 = np.array(data_branch2)
    X_branch2 = X_branch2.reshape(len(X_branch2), 90, 3)
    return X_branch2


def load_branch3_data_HTCas9(filename="Feature_MLP_expanded_HTCas9.txt"):
    """
    Loads 120-dimensional numeric features from file.
    Each line has 120 floats (space- or comma-delimited).
    """
    X_branch3 = []
    with open(filename, 'r') as file:
        for line in file:
            line = line.strip()
            # Expect 120 numbers per line
            values = line.replace('[', '').replace(']', '').split()
            float_vals = [float(v.strip().replace(',', '')) for v in values]
            X_branch3.append(float_vals)
    
    X_branch3 = np.array(X_branch3)  # shape: (n_samples, 120)
    X_branch3 = X_branch3.reshape(len(X_branch3), 120)
    return X_branch3

def load_branch3_data_HT11(filename="Feature_MLP_expanded_HT11.txt"):
    """
    Loads 120-dimensional numeric features from file.
    Each line has 120 floats (space- or comma-delimited).
    """
    X_branch3 = []
    with open(filename, 'r') as file:
        for line in file:
            line = line.strip()
            # Expect 120 numbers per line
            values = line.replace('[', '').replace(']', '').split()
            float_vals = [float(v.strip().replace(',', '')) for v in values]
            X_branch3.append(float_vals)
    
    X_branch3 = np.array(X_branch3)  # shape: (n_samples, 120)
    X_branch3 = X_branch3.reshape(len(X_branch3), 120)
    return X_branch3

def load_branch3_data_mmc4(filename="Feature_MLP_expanded_mmc4.txt"):
    """
    Loads 120-dimensional numeric features from file.
    Each line has 120 floats (space- or comma-delimited).
    """
    X_branch3 = []
    with open(filename, 'r') as file:
        for line in file:
            line = line.strip()
            # Expect 120 numbers per line
            values = line.replace('[', '').replace(']', '').split()
            float_vals = [float(v.strip().replace(',', '')) for v in values]
            X_branch3.append(float_vals)
    
    X_branch3 = np.array(X_branch3)  # shape: (n_samples, 120)
    X_branch3 = X_branch3.reshape(len(X_branch3), 120)
    return X_branch3

def load_branch3_data_mmc5(filename="Feature_MLP_expanded_mmc5.txt"):
    """
    Loads 120-dimensional numeric features from file.
    Each line has 120 floats (space- or comma-delimited).
    """
    X_branch3 = []
    with open(filename, 'r') as file:
        for line in file:
            line = line.strip()
            # Expect 120 numbers per line
            values = line.replace('[', '').replace(']', '').split()
            float_vals = [float(v.strip().replace(',', '')) for v in values]
            X_branch3.append(float_vals)
    
    X_branch3 = np.array(X_branch3)  # shape: (n_samples, 120)
    X_branch3 = X_branch3.reshape(len(X_branch3), 120)
    return X_branch3

def load_branch3_data_change_seq(filename="Feature_MLP_expanded_change_seq.txt"):
    """
    Loads 120-dimensional numeric features from file.
    Each line has 120 floats (space- or comma-delimited).
    """
    X_branch3 = []
    with open(filename, 'r') as file:
        for line in file:
            line = line.strip()
            # Expect 120 numbers per line
            values = line.replace('[', '').replace(']', '').split()
            float_vals = [float(v.strip().replace(',', '')) for v in values]
            X_branch3.append(float_vals)
    
    X_branch3 = np.array(X_branch3)  # shape: (n_samples, 120)
    X_branch3 = X_branch3.reshape(len(X_branch3), 120)
    return X_branch3

def load_branch3_data_tiger(filename="Feature_MLP_expanded_tiger.txt"):
    """
    Loads 120-dimensional numeric features from file.
    Each line has 120 floats (space- or comma-delimited).
    """
    X_branch3 = []
    with open(filename, 'r') as file:
        for line in file:
            line = line.strip()
            # Expect 120 numbers per line
            values = line.replace('[', '').replace(']', '').split()
            float_vals = [float(v.strip().replace(',', '')) for v in values]
            X_branch3.append(float_vals)
    
    X_branch3 = np.array(X_branch3)  # shape: (n_samples, 120)
    X_branch3 = X_branch3.reshape(len(X_branch3), 120)
    return X_branch3

def load_branch3_data_iMeta(filename="Feature_MLP_expanded_iMeta.txt"):
    """
    Loads 120-dimensional numeric features from file.
    Each line has 120 floats (space- or comma-delimited).
    """
    X_branch3 = []
    with open(filename, 'r') as file:
        for line in file:
            line = line.strip()
            # Expect 120 numbers per line
            values = line.replace('[', '').replace(']', '').split()
            float_vals = [float(v.strip().replace(',', '')) for v in values]
            X_branch3.append(float_vals)
    
    X_branch3 = np.array(X_branch3)  # shape: (n_samples, 120)
    X_branch3 = X_branch3.reshape(len(X_branch3), 120)
    return X_branch3


def load_reaction_rates_HTCas9():
    with open('HTCas9_indel_frequency_value_percentage_unique.txt', 'r') as f:
        rates = [float(line.strip()) for line in f.readlines()]
    return np.array(rates)

def load_reaction_rates_HT11():
    with open('HT1-1_indel_frequency_value_percentage_TTTV_filtered.txt', 'r') as f:
        rates = [float(line.strip()) for line in f.readlines()]
    return np.array(rates)

def load_reaction_rates_mmc4():
    with open('filtered_ratio_NM_002794_4.txt', 'r') as f:
        rates = [float(line.strip()) for line in f.readlines()]
    return np.array(rates)

def load_reaction_rates_mmc5():
    with open('RfxCas13d_validation_expression_level_combined.txt', 'r') as f:
        rates = [float(line.strip()) for line in f.readlines()]
    return np.array(rates)

def load_reaction_rates_change_seq():
    with open('normalized_change_seq_log_transformed_20nt_sampled_2.txt', 'r') as f:
        rates = [float(line.strip()) for line in f.readlines()]
    return np.array(rates)

def load_reaction_rates_tiger():
    with open('normalized_off_target_log2_fold_change_d15_values_EIF3B_filtered_sampled_1.txt', 'r') as f:
        rates = [float(line.strip()) for line in f.readlines()]
    return np.array(rates)

def load_reaction_rates_iMeta():
    with open('normalized_iMeta_trans_LbCas12a_sorted_data_30min_TTTV_filtered.txt', 'r') as f:
        rates = [float(line.strip()) for line in f.readlines()]
    return np.array(rates)


########################################
# 2. Graph Data Utilities (for GNN)
########################################

def matrix_to_edge_index(prob_matrix):
    edges_source = []
    edges_target = []
    L = len(prob_matrix)
    for i in range(L):
        for j in range(i + 1, L):  # only i < j
            if prob_matrix[i][j] > 0.01:
                edges_source.append(i)
                edges_target.append(j)
    return [edges_source, edges_target]

def matrix_to_edge_index_collapsed(prob_matrix: np.ndarray, threshold: float = 0.01):
    """
    Build undirected edges (i < j) from a dense probability matrix using a threshold.
    Returns edge_index as [sources, targets] lists of ints.
    """
    src, dst = [], []
    L = prob_matrix.shape[0]
    for i in range(L):
        for j in range(i+1, L):
            if prob_matrix[i, j] > threshold:
                src.append(i)
                dst.append(j)
    return [src, dst]

def generate_edge_features(edge_index, prob_matrix):
    if not edge_index:
        return []
    feats = []
    for i in range(len(edge_index[0])):
        src = edge_index[0][i]; tgt = edge_index[1][i]
        feats.append([prob_matrix[src][tgt]])  # shape [E, 1]
    return feats

def _num_nodes_from_edge_index(edge_index_list, fallback):
    if not edge_index_list or len(edge_index_list[0]) == 0:
        return fallback
    mx = 0
    for a in edge_index_list:
        if len(a) > 0:
            mx = max(mx, max(a))
    return mx + 1


def collapse_to_first_k_plus_1_sum_HTCas9(prob_matrix: np.ndarray, first_k: int = 20) -> np.ndarray:
    """
    Collapse an LxL base-pair probability matrix into (first_k + 1) x (first_k + 1):
      - Rows/cols 0..first_k-1: original 1..first_k bases
      - Row/col first_k: an 'outside' super-node representing bases >= first_k
    For each i in [0..first_k-1], the edge prob to the super-node is the
    SUM of probabilities of pairing to bases >= first_k.
    """
    L = prob_matrix.shape[0]
    K = first_k
    new_size = K + 1
    newP = np.zeros((new_size, new_size), dtype=float)

    # Copy the 0..K-1 block
    newP[:K, :K] = prob_matrix[:K, :K][::-1, ::-1]

    if L > K:
        # Outside block (K..L-1)
        outside_block = prob_matrix[:K, K:L]  # shape: K x (L-K)
        # Sum of probabilities across outside bases
        p_sum = np.sum(outside_block, axis=1)
        # Fill symmetric connections to the super-node K
        newP[:K, K] = p_sum
        newP[K, :K] = p_sum

    # Leave newP[K, K] as 0
    return newP

def collapse_to_last_k_plus_1_sum_HT11(prob_matrix: np.ndarray, last_k: int = 20) -> np.ndarray:
    """
    last_k refers to the spacer length
    Collapse an LxL base-pair probability matrix into (last_k + 1) x (last_k + 1):
      - Rows/cols last_k..39: original last_k_base..40
      - Row/col 0: last_k: an 'outside' super-node representing bases <= last_k
    For each i in [last_k..39], the edge prob to the super-node is the
    SUM of probabilities of pairing to bases <= last_k.
    """
    L = prob_matrix.shape[0]
    K = last_k
    new_size = K + 1
    newP = np.zeros((new_size, new_size), dtype=float)

    # Copy the K..39 block
    newP[:K, :K] = prob_matrix[K:, K:]

    if L > K:
        # Outside block (0..K-1)
        outside_block = prob_matrix[K:L, :K]  # shape: K x (L-K)
        # Sum of probabilities across outside bases
        p_sum = np.sum(outside_block, axis=1)
        # Fill symmetric connections to the super-node K
        newP[:K, K] = p_sum
        newP[K, :K] = p_sum

    # Leave newP[K, K] as 0
    return newP

def collapse_to_last_k_plus_1_sum_RfxCas13d(prob_matrix: np.ndarray, last_k: int = 30) -> np.ndarray:
    """
    last_k refers to the spacer length
    Collapse an LxL base-pair probability matrix into (last_k + 1) x (last_k + 1):
      - Rows/cols last_k..65: original last_k_base..66
      - Row/col 0: last_k: an 'outside' super-node representing bases <= last_k
    For each i in [last_k..65], the edge prob to the super-node is the
    SUM of probabilities of pairing to bases <= last_k.
    """
    L = prob_matrix.shape[0]
    K = last_k
    new_size = K + 1
    newP = np.zeros((new_size, new_size), dtype=float)

    # Copy the K..65 block
    newP[:K, :K] = prob_matrix[(L-K):, (L-K):]

    if L > K:
        # Outside block (0..K-1)
        outside_block = prob_matrix[(L-K):L, :(L-K)]  # shape: K x (L-K)
        # Sum of probabilities across outside bases
        p_sum = np.sum(outside_block, axis=1)
        # Fill symmetric connections to the super-node K
        newP[:K, K] = p_sum
        newP[K, :K] = p_sum

    # Leave newP[K, K] as 0
    return newP

def collapse_to_last_k_plus_1_sum_tiger(prob_matrix: np.ndarray, last_k: int = 23) -> np.ndarray:
    """
    last_k refers to the spacer length
    Collapse an LxL base-pair probability matrix into (last_k + 1) x (last_k + 1):
      - Rows/cols last_k..52: original last_k_base..53
      - Row/col 0: last_k: an 'outside' super-node representing bases <= last_k
    For each i in [last_k..52], the edge prob to the super-node is the
    SUM of probabilities of pairing to bases <= last_k.
    """
    L = prob_matrix.shape[0]
    K = last_k
    new_size = K + 1
    newP = np.zeros((new_size, new_size), dtype=float)

    # Copy the K..52 block
    newP[:K, :K] = prob_matrix[(L-K):, (L-K):]

    if L > K:
        # Outside block (0..K-1)
        outside_block = prob_matrix[(L-K):L, :(L-K)]  # shape: K x (L-K)
        # Sum of probabilities across outside bases
        p_sum = np.sum(outside_block, axis=1)
        # Fill symmetric connections to the super-node K
        newP[:K, K] = p_sum
        newP[K, :K] = p_sum

    # Leave newP[K, K] as 0
    return newP

def collapse_to_last_k_plus_1_sum_iMeta(prob_matrix: np.ndarray, last_k: int = 21) -> np.ndarray:
    """
    last_k refers to the spacer length
    Collapse an LxL base-pair probability matrix into (last_k + 1) x (last_k + 1):
      - Rows/cols last_k..41: original last_k_base..42
      - Row/col 0: last_k: an 'outside' super-node representing bases <= last_k
    For each i in [last_k..41], the edge prob to the super-node is the
    SUM of probabilities of pairing to bases <= last_k.
    """
    L = prob_matrix.shape[0]
    K = last_k
    new_size = K + 1
    newP = np.zeros((new_size, new_size), dtype=float)

    # Copy the K..53 block
    newP[:K, :K] = prob_matrix[(L-K):, (L-K):]

    if L > K:
        # Outside block (0..K-1)
        outside_block = prob_matrix[(L-K):L, :(L-K)]  # shape: K x (L-K)
        # Sum of probabilities across outside bases
        p_sum = np.sum(outside_block, axis=1)
        # Fill symmetric connections to the super-node K
        newP[:K, K] = p_sum
        newP[K, :K] = p_sum

    # Leave newP[K, K] as 0
    return newP


def slice_and_renumber_duplex_HTCas9(prob_matrix, len_guide, len_target, k=20):
    """
    Keep only the first k positions of the guide and the last k positions of the target,
    then reverse the order within each block (guide: k→1, target: 2k→k+1).
    Returns a (kg+kt) x (kg+kt) reduced/renumbered matrix.
    """
    L = len_guide + len_target
    assert prob_matrix.shape == (L, L), "prob_matrix size mismatch."

    kg = min(k, len_guide)
    kt = min(k, len_target)

    # Original indices: guide = [0..len_guide-1], target = [len_guide..L-1]
    sel_g = np.arange(0, kg)                      # first k of guide
    sel_t = np.arange(L - kt, L)                  # last k of target
    S = np.concatenate([sel_g, sel_t])

    # Extract submatrix for these rows/cols
    P = prob_matrix[np.ix_(S, S)].copy()

    # Reverse guide rows/cols
    P[:kg, :] = P[:kg, :][::-1, :]
    P[:, :kg] = P[:, :kg][:, ::-1]

    return P

def slice_and_renumber_duplex_HT11(prob_matrix, len_guide, len_target, k=20):
    """
    k refers to the spacer length
    Keep only the last k positions of the guide and the last k positions of the target,
    Returns a (kg+kt) x (kg+kt) reduced/renumbered matrix.
    """
    L = len_guide + len_target
    assert prob_matrix.shape == (L, L), "prob_matrix size mismatch."

    kg = min(k, len_guide)
    kt = min(k, len_target)

    # Original indices: guide = [0..len_guide-1], target = [len_guide..L-1]
    sel_g = np.arange(len_guide - kg, len_guide)  # last k of guide
    sel_t = np.arange(L - kt, L)                  # last k of target
    S = np.concatenate([sel_g, sel_t])

    # Extract submatrix for these rows/cols
    P = prob_matrix[np.ix_(S, S)].copy()

    # Reverse target rows/cols
    P[kg:, :] = P[kg:, :][::-1, :]
    P[:, kg:] = P[:, kg:][:, ::-1]

    return P

def slice_and_renumber_duplex_RfxCas13d(prob_matrix, len_guide, len_target, k=30):
    """
    k refers to the spacer length
    Keep only the last k positions of the guide and the last k positions of the target,
    Returns a (kg+kt) x (kg+kt) reduced/renumbered matrix.
    """
    L = len_guide + len_target
    assert prob_matrix.shape == (L, L), "prob_matrix size mismatch."

    kg = min(k, len_guide)
    kt = min(k, len_target)

    # Original indices: guide = [0..len_guide-1], target = [len_guide..L-1]
    sel_g = np.arange(len_guide - kg, len_guide)  # last k of guide
    sel_t = np.arange(L - kt, L)                  # last k of target
    S = np.concatenate([sel_g, sel_t])

    # Extract submatrix for these rows/cols
    P = prob_matrix[np.ix_(S, S)].copy()

    # Reverse target rows/cols
    P[kg:, :] = P[kg:, :][::-1, :]
    P[:, kg:] = P[:, kg:][:, ::-1]

    return P

def slice_and_renumber_duplex_tiger(prob_matrix, len_guide, len_target, k=23):
    """
    k refers to the spacer length
    Keep only the last k positions of the guide and the last k positions of the target,
    Returns a (kg+kt) x (kg+kt) reduced/renumbered matrix.
    """
    L = len_guide + len_target
    assert prob_matrix.shape == (L, L), "prob_matrix size mismatch."

    kg = min(k, len_guide)
    kt = min(k, len_target)

    # Original indices: guide = [0..len_guide-1], target = [len_guide..L-1]
    sel_g = np.arange(len_guide - kg, len_guide)  # last k of guide
    sel_t = np.arange(L - kt, L)                  # last k of target
    S = np.concatenate([sel_g, sel_t])

    # Extract submatrix for these rows/cols
    P = prob_matrix[np.ix_(S, S)].copy()

    # Reverse target rows/cols
    P[kg:, :] = P[kg:, :][::-1, :]
    P[:, kg:] = P[:, kg:][:, ::-1]

    return P

def slice_and_renumber_duplex_iMeta(prob_matrix, len_guide, len_target, k=21):
    """
    k refers to the spacer length
    Keep only the last k positions of the guide and the last k positions of the target,
    Returns a (kg+kt) x (kg+kt) reduced/renumbered matrix.
    """
    L = len_guide + len_target
    assert prob_matrix.shape == (L, L), "prob_matrix size mismatch."

    kg = min(k, len_guide)
    kt = min(k, len_target)

    # Original indices: guide = [0..len_guide-1], target = [len_guide..L-1]
    sel_g = np.arange(len_guide - kg, len_guide)  # last k of guide
    sel_t = np.arange(L - kt, L)                  # last k of target
    S = np.concatenate([sel_g, sel_t])

    # Extract submatrix for these rows/cols
    P = prob_matrix[np.ix_(S, S)].copy()

    # Reverse target rows/cols
    P[kg:, :] = P[kg:, :][::-1, :]
    P[:, kg:] = P[:, kg:][:, ::-1]

    return P


# NUPACK models
from nupack import Model, pairs
my_model_RNA = Model(material='rna', ensemble='nostacking', celsius=37, sodium=0.1, magnesium=0.01)
my_model_DNA = Model(material='dna', ensemble='nostacking', celsius=37, sodium=0.1, magnesium=0.01)

def DNA_reverse_complement(DNA):
    complement = {'A': 'T', 'C': 'G', 'G': 'C', 'T': 'A'}
    return ''.join(complement.get(base, base) for base in reversed(DNA))

def load_dna_reaction_data_HTCas9():
    with open('HTCas9_full_guide_sequences_unique.txt', 'r') as f:
        crna_sequences = [line.strip() for line in f.readlines()]
    with open('HTCas9_target_sequences_noPAM_unique.txt', 'r') as f:
        target_sequences = [line.strip() for line in f.readlines()]

    FIRST_K = 20  # keep first 20 bases, collapse the rest into node index 20 (21st base)
        
    crna_edge_indices = []
    crna_edge_features = []
    for guide in crna_sequences:
        prob_full = pairs(strands=guide, model=my_model_RNA).to_array()
        prob_collapsed = collapse_to_first_k_plus_1_sum_HTCas9(prob_full, first_k=FIRST_K)
        edge_index = matrix_to_edge_index_collapsed(prob_collapsed, threshold=0.01)
        feats = generate_edge_features(edge_index, prob_collapsed)
        crna_edge_indices.append(edge_index)   # each is [sources, targets] with 0..20 node indices
        crna_edge_features.append(feats)       # [[p_ij], ...]

    ssDNA_target_bh_edge_indices = []
    ssDNA_target_bh_edge_features = []
    for i in range (0, len(crna_sequences)):
        prob_matrix = pairs(strands=target_sequences[i], model=my_model_DNA).to_array()
        edge_index = matrix_to_edge_index(prob_matrix)
        feats = generate_edge_features(edge_index, prob_matrix)
        ssDNA_target_bh_edge_indices.append(edge_index)
        ssDNA_target_bh_edge_features.append(feats)
    
    duplex_edge_indices = []
    duplex_edge_features = []
    
    for i in range (0, len(crna_sequences)):
        prob_matrix = pairs(strands=[crna_sequences[i], target_sequences[i]], model=my_model_RNA).to_array()
        len_g, len_t = len(crna_sequences[i]), len(target_sequences[i])
        # Crop to guide[1..20] + target[last 20], renumber both reversed
        updated_P = slice_and_renumber_duplex_HTCas9(prob_matrix, len_g, len_t, k=20)
        edge_index = matrix_to_edge_index(updated_P)
        feats = generate_edge_features(edge_index, updated_P)
        duplex_edge_indices.append(edge_index)
        duplex_edge_features.append(feats)

    
    with open('HTCas9_indel_frequency_value_percentage_unique.txt', 'r') as f:
        reaction_rates = [float(line.strip()) for line in f.readlines()]
    
    return (crna_edge_indices, ssDNA_target_bh_edge_indices, duplex_edge_indices, 
            crna_edge_features, ssDNA_target_bh_edge_features, duplex_edge_features, 
            reaction_rates)

def load_dna_reaction_data_HT11():
    with open('HT1-1_full_guide_sequences_TTTV_filtered.txt', 'r') as f:
        crna_sequences = [line.strip() for line in f.readlines()]
    with open('HT1-1_target_sequences_noPAM_TTTV_filtered.txt', 'r') as f:
        target_sequences = [line.strip() for line in f.readlines()]

    LAST_K = 20  # keep last 20 bases, collapse the rest into node index 20 (21st base)
        
    crna_edge_indices = []
    crna_edge_features = []
    for guide in crna_sequences:
        prob_full = pairs(strands=guide, model=my_model_RNA).to_array()
        prob_collapsed = collapse_to_last_k_plus_1_sum_HT11(prob_full, last_k=LAST_K)
        edge_index = matrix_to_edge_index_collapsed(prob_collapsed, threshold=0.01)
        feats = generate_edge_features(edge_index, prob_collapsed)
        crna_edge_indices.append(edge_index)   # each is [sources, targets] with 0..20 node indices
        crna_edge_features.append(feats)       # [[p_ij], ...]

    ssDNA_target_bh_edge_indices = []
    ssDNA_target_bh_edge_features = []
    for i in range (0, len(crna_sequences)):
        prob_matrix = pairs(strands=target_sequences[i], model=my_model_DNA).to_array()
        prob_matrix = prob_matrix[::-1, ::-1]
        edge_index = matrix_to_edge_index(prob_matrix)
        feats = generate_edge_features(edge_index, prob_matrix)
        ssDNA_target_bh_edge_indices.append(edge_index)
        ssDNA_target_bh_edge_features.append(feats)
    
    duplex_edge_indices = []
    duplex_edge_features = []
    
    for i in range (0, len(crna_sequences)):
        prob_matrix = pairs(strands=[crna_sequences[i], target_sequences[i]], model=my_model_RNA).to_array()
        len_g, len_t = len(crna_sequences[i]), len(target_sequences[i])
        # Crop to guide[20..40] + target[last 20], renumber both
        updated_P = slice_and_renumber_duplex_HT11(prob_matrix, len_g, len_t, k=20)
        edge_index = matrix_to_edge_index(updated_P)
        feats = generate_edge_features(edge_index, updated_P)
        duplex_edge_indices.append(edge_index)
        duplex_edge_features.append(feats)

    
    with open('HT1-1_indel_frequency_value_percentage_TTTV_filtered.txt', 'r') as f:
        reaction_rates = [float(line.strip()) for line in f.readlines()]
    
    return (crna_edge_indices, ssDNA_target_bh_edge_indices, duplex_edge_indices, 
            crna_edge_features, ssDNA_target_bh_edge_features, duplex_edge_features, 
            reaction_rates)


def load_rna_reaction_data_mmc4():
    with open('RfxCas13d_mmc4_filtered_full_guide_sequences_NM_002794_4.txt', 'r') as f:
        crna_sequences = [line.strip() for line in f.readlines()]
    with open('RfxCas13d_mmc4_filtered_target_sequences_noPAM_NM_002794_4.txt', 'r') as f:
        target_sequences = [line.strip() for line in f.readlines()]

    LAST_K = 30  # keep last 30 bases, collapse the rest into node index 30 (31st base)
        
    crna_edge_indices = []
    crna_edge_features = []
    for guide in crna_sequences:
        prob_full = pairs(strands=guide, model=my_model_RNA).to_array()
        prob_collapsed = collapse_to_last_k_plus_1_sum_RfxCas13d(prob_full, last_k=LAST_K)
        edge_index = matrix_to_edge_index_collapsed(prob_collapsed, threshold=0.01)
        feats = generate_edge_features(edge_index, prob_collapsed)
        crna_edge_indices.append(edge_index)   # each is [sources, targets] with 0..30 node indices
        crna_edge_features.append(feats)       # [[p_ij], ...]

    ssRNA_target_bh_edge_indices = []
    ssRNA_target_bh_edge_features = []
    for i in range (0, len(crna_sequences)):
        prob_matrix = pairs(strands=target_sequences[i], model=my_model_RNA).to_array()
        prob_matrix = prob_matrix[::-1, ::-1]
        edge_index = matrix_to_edge_index(prob_matrix)
        feats = generate_edge_features(edge_index, prob_matrix)
        ssRNA_target_bh_edge_indices.append(edge_index)
        ssRNA_target_bh_edge_features.append(feats)

    
    duplex_edge_indices = []
    duplex_edge_features = []
    
    for i in range (0, len(crna_sequences)):
        prob_matrix = pairs(strands=[crna_sequences[i], target_sequences[i]], model=my_model_RNA).to_array()
        len_g, len_t = len(crna_sequences[i]), len(target_sequences[i])
        # Crop to guide[36..66] + target[last 30], renumber both
        updated_P = slice_and_renumber_duplex_RfxCas13d(prob_matrix, len_g, len_t, k=30)
        edge_index = matrix_to_edge_index(updated_P)
        feats = generate_edge_features(edge_index, updated_P)
        duplex_edge_indices.append(edge_index)
        duplex_edge_features.append(feats)

    
    with open('filtered_ratio_NM_002794_4.txt', 'r') as f:
        reaction_rates = [float(line.strip()) for line in f.readlines()]
    
    return (crna_edge_indices, ssRNA_target_bh_edge_indices, duplex_edge_indices, 
            crna_edge_features, ssRNA_target_bh_edge_features, duplex_edge_features, 
            reaction_rates)


def load_rna_reaction_data_mmc5():
    with open('RfxCas13d_validation_full_guide_sequences_combined.txt', 'r') as f:
        crna_sequences = [line.strip() for line in f.readlines()]
    with open('RfxCas13d_validation_target_sequences_noPAM_combined.txt', 'r') as f:
        target_sequences = [line.strip() for line in f.readlines()]

    LAST_K = 30  # keep last 30 bases, collapse the rest into node index 30 (31st base)
        
    crna_edge_indices = []
    crna_edge_features = []
    for guide in crna_sequences:
        prob_full = pairs(strands=guide, model=my_model_RNA).to_array()
        prob_collapsed = collapse_to_last_k_plus_1_sum_RfxCas13d(prob_full, last_k=LAST_K)
        edge_index = matrix_to_edge_index_collapsed(prob_collapsed, threshold=0.01)
        feats = generate_edge_features(edge_index, prob_collapsed)
        crna_edge_indices.append(edge_index)   # each is [sources, targets] with 0..30 node indices
        crna_edge_features.append(feats)       # [[p_ij], ...]

    ssRNA_target_bh_edge_indices = []
    ssRNA_target_bh_edge_features = []
    for i in range (0, len(crna_sequences)):
        prob_matrix = pairs(strands=target_sequences[i], model=my_model_RNA).to_array()
        prob_matrix = prob_matrix[::-1, ::-1]
        edge_index = matrix_to_edge_index(prob_matrix)
        feats = generate_edge_features(edge_index, prob_matrix)
        ssRNA_target_bh_edge_indices.append(edge_index)
        ssRNA_target_bh_edge_features.append(feats)

    duplex_edge_indices = []
    duplex_edge_features = []
    
    for i in range (0, len(crna_sequences)):
        prob_matrix = pairs(strands=[crna_sequences[i], target_sequences[i]], model=my_model_RNA).to_array()
        len_g, len_t = len(crna_sequences[i]), len(target_sequences[i])
        # Crop to guide[36..66] + target[last 30], renumber both
        updated_P = slice_and_renumber_duplex_RfxCas13d(prob_matrix, len_g, len_t, k=30)
        edge_index = matrix_to_edge_index(updated_P)
        feats = generate_edge_features(edge_index, updated_P)
        duplex_edge_indices.append(edge_index)
        duplex_edge_features.append(feats)
    
    with open('RfxCas13d_validation_expression_level_combined.txt', 'r') as f:
        reaction_rates = [float(line.strip()) for line in f.readlines()]
    
    return (crna_edge_indices, ssRNA_target_bh_edge_indices, duplex_edge_indices, 
            crna_edge_features, ssRNA_target_bh_edge_features, duplex_edge_features, 
            reaction_rates)


def load_dna_reaction_data_change_seq():
    with open('change_seq_full_guide_sequences_20nt_sampled_2.txt', 'r') as f:
        crna_sequences = [line.strip() for line in f.readlines()]
    with open('change_seq_target_sequences_noPAM_20nt_sampled_2.txt', 'r') as f:
        target_sequences = [line.strip() for line in f.readlines()]

    FIRST_K = 20  # keep first 20 bases, collapse the rest into node index 20 (21st base)
        
    crna_edge_indices = []
    crna_edge_features = []
    for guide in crna_sequences:
        prob_full = pairs(strands=guide, model=my_model_RNA).to_array()
        prob_collapsed = collapse_to_first_k_plus_1_sum_HTCas9(prob_full, first_k=FIRST_K)
        edge_index = matrix_to_edge_index_collapsed(prob_collapsed, threshold=0.01)
        feats = generate_edge_features(edge_index, prob_collapsed)
        crna_edge_indices.append(edge_index)   # each is [sources, targets] with 0..20 node indices
        crna_edge_features.append(feats)       # [[p_ij], ...]

    ssDNA_target_bh_edge_indices = []
    ssDNA_target_bh_edge_features = []
    for i in range (0, len(crna_sequences)):
        prob_matrix = pairs(strands=target_sequences[i], model=my_model_DNA).to_array()
        edge_index = matrix_to_edge_index(prob_matrix)
        feats = generate_edge_features(edge_index, prob_matrix)
        ssDNA_target_bh_edge_indices.append(edge_index)
        ssDNA_target_bh_edge_features.append(feats)

    
    duplex_edge_indices = []
    duplex_edge_features = []
    
    for i in range (0, len(crna_sequences)):
        prob_matrix = pairs(strands=[crna_sequences[i], target_sequences[i]], model=my_model_RNA).to_array()
        len_g, len_t = len(crna_sequences[i]), len(target_sequences[i])
        # Crop to guide[1..20] + target[last 20], renumber both reversed
        updated_P = slice_and_renumber_duplex_HTCas9(prob_matrix, len_g, len_t, k=20)
        edge_index = matrix_to_edge_index(updated_P)
        feats = generate_edge_features(edge_index, updated_P)
        duplex_edge_indices.append(edge_index)
        duplex_edge_features.append(feats)

    
    with open('normalized_change_seq_log_transformed_20nt_sampled_2.txt', 'r') as f:
        reaction_rates = [float(line.strip()) for line in f.readlines()]
    
    return (crna_edge_indices, ssDNA_target_bh_edge_indices, duplex_edge_indices, 
            crna_edge_features, ssDNA_target_bh_edge_features, duplex_edge_features, 
            reaction_rates)

def load_rna_reaction_data_tiger():
    with open('off_target_full_guide_spacer_sequences_EIF3B_filtered_sampled_1.txt', 'r') as f:
        crna_sequences = [line.strip() for line in f.readlines()]
    with open('off_target_noPAM_target_sequences_EIF3B_filtered_sampled_1.txt', 'r') as f:
        target_sequences = [line.strip() for line in f.readlines()]

    LAST_K = 23  # keep last 23 bases, collapse the rest into node index 23 (24st base)
        
    crna_edge_indices = []
    crna_edge_features = []
    for guide in crna_sequences:
        prob_full = pairs(strands=guide, model=my_model_RNA).to_array()
        prob_collapsed = collapse_to_last_k_plus_1_sum_tiger(prob_full, last_k=LAST_K)
        edge_index = matrix_to_edge_index_collapsed(prob_collapsed, threshold=0.01)
        feats = generate_edge_features(edge_index, prob_collapsed)
        crna_edge_indices.append(edge_index)   # each is [sources, targets] with 0..23 node indices
        crna_edge_features.append(feats)       # [[p_ij], ...]

    ssRNA_target_bh_edge_indices = []
    ssRNA_target_bh_edge_features = []
    for i in range (0, len(crna_sequences)):
        prob_matrix = pairs(strands=target_sequences[i], model=my_model_RNA).to_array()
        prob_matrix = prob_matrix[::-1, ::-1]
        edge_index = matrix_to_edge_index(prob_matrix)
        feats = generate_edge_features(edge_index, prob_matrix)
        ssRNA_target_bh_edge_indices.append(edge_index)
        ssRNA_target_bh_edge_features.append(feats)

    duplex_edge_indices = []
    duplex_edge_features = []

    
    for i in range (0, len(crna_sequences)):
        prob_matrix = pairs(strands=[crna_sequences[i], target_sequences[i]], model=my_model_RNA).to_array()
        len_g, len_t = len(crna_sequences[i]), len(target_sequences[i])
        # Crop to guide[30..53] + target[last 23], renumber both
        updated_P = slice_and_renumber_duplex_tiger(prob_matrix, len_g, len_t, k=23)
        edge_index = matrix_to_edge_index(updated_P)
        feats = generate_edge_features(edge_index, updated_P)
        duplex_edge_indices.append(edge_index)
        duplex_edge_features.append(feats)
    
    with open('normalized_off_target_log2_fold_change_d15_values_EIF3B_filtered_sampled_1.txt', 'r') as f:
        reaction_rates = [float(line.strip()) for line in f.readlines()]
    
    return (crna_edge_indices, ssRNA_target_bh_edge_indices, duplex_edge_indices, 
            crna_edge_features, ssRNA_target_bh_edge_features, duplex_edge_features, 
            reaction_rates)


def load_dna_reaction_data_iMeta():
    with open('iMeta_trans_LbCas12a_sorted_guide_TTTV_filtered.txt', 'r') as f:
        crna_sequences = [line.strip() for line in f.readlines()]
    with open('iMeta_trans_LbCas12a_sorted_target_noPAM_TTTV_filtered.txt', 'r') as f:
        target_sequences = [line.strip() for line in f.readlines()]

    LAST_K = 21  # keep last 21 bases, collapse the rest into node index 21 (22nd base)
        
    crna_edge_indices = []
    crna_edge_features = []
    for guide in crna_sequences:
        prob_full = pairs(strands=guide, model=my_model_RNA).to_array()
        prob_collapsed = collapse_to_last_k_plus_1_sum_iMeta(prob_full, last_k=LAST_K)
        edge_index = matrix_to_edge_index_collapsed(prob_collapsed, threshold=0.01)
        feats = generate_edge_features(edge_index, prob_collapsed)
        crna_edge_indices.append(edge_index)   # each is [sources, targets] with 0..21 node indices
        crna_edge_features.append(feats)       # [[p_ij], ...]

    ssDNA_target_bh_edge_indices = []
    ssDNA_target_bh_edge_features = []
    for i in range (0, len(crna_sequences)):
        prob_matrix = pairs(strands=target_sequences[i], model=my_model_DNA).to_array()
        prob_matrix = prob_matrix[::-1, ::-1]
        edge_index = matrix_to_edge_index(prob_matrix)
        feats = generate_edge_features(edge_index, prob_matrix)
        ssDNA_target_bh_edge_indices.append(edge_index)
        ssDNA_target_bh_edge_features.append(feats)

    
    duplex_edge_indices = []
    duplex_edge_features = []
    
    for i in range (0, len(crna_sequences)):
        prob_matrix = pairs(strands=[crna_sequences[i], target_sequences[i]], model=my_model_RNA).to_array()
        len_g, len_t = len(crna_sequences[i]), len(target_sequences[i])
        # Crop to guide[21..42] + target[last 21], renumber both
        updated_P = slice_and_renumber_duplex_iMeta(prob_matrix, len_g, len_t, k=21)
        edge_index = matrix_to_edge_index(updated_P)
        feats = generate_edge_features(edge_index, updated_P)
        duplex_edge_indices.append(edge_index)
        duplex_edge_features.append(feats)

    with open('normalized_iMeta_trans_LbCas12a_sorted_data_30min_TTTV_filtered.txt', 'r') as f:
        reaction_rates = [float(line.strip()) for line in f.readlines()]
    
    return (crna_edge_indices, ssDNA_target_bh_edge_indices, duplex_edge_indices, 
            crna_edge_features, ssDNA_target_bh_edge_features, duplex_edge_features, 
            reaction_rates)


from typing import Iterable, List, Tuple, Any

# Each graph set is a tuple:
# (crna_ei, tbh_ei, d_ei, crna_ea, tbh_ea, d_ea, y)
GraphSet = Tuple[Iterable[Any], Iterable[Any], Iterable[Any],
                 Iterable[Any], Iterable[Any], Iterable[Any], Iterable[Any]]

def combine_dna_graph_sets(*graph_sets: GraphSet):
    """Concatenate any number of DNA/RNA graph sets along the sample axis.
       Optionally return a source label per sample (0,1,2,...) indicating origin."""
    if not graph_sets:
        raise ValueError("Provide at least one graph set")

    c_ei: List[Any] = []
    tbh_ei: List[Any] = []
    d_ei: List[Any] = []
    c_ea: List[Any] = []
    tbh_ea: List[Any] = []
    d_ea: List[Any] = []
    y: List[Any] = []

    for sid, gs in enumerate(graph_sets):
        try:
            c_ei_a, tbh_ei_a, d_ei_a, c_ea_a, tbh_ea_a, d_ea_a, y_a = gs
        except Exception as e:
            raise ValueError(f"Graph set #{sid} must be a 7-tuple") from e

        # Optional consistency check per set
        n = len(y_a)
        if not all(len(lst) == n for lst in (c_ei_a, tbh_ei_a, d_ei_a, c_ea_a, tbh_ea_a, d_ea_a)):
            raise ValueError(f"Graph set #{sid} has mismatched lengths")

        c_ei.extend(list(c_ei_a))
        tbh_ei.extend(list(tbh_ei_a))
        d_ei.extend(list(d_ei_a))
        c_ea.extend(list(c_ea_a))
        tbh_ea.extend(list(tbh_ea_a))
        d_ea.extend(list(d_ea_a))
        y.extend(list(y_a))

    combined = (c_ei, tbh_ei, d_ei, c_ea, tbh_ea, d_ea, y)
    return combined



class CRNADuplexSubstructuresDataset(Dataset):
    def __init__(self, crna_edge_indices, ssDNA_target_bh_edge_indices, duplex_edge_indices, 
                 crna_edge_features, ssDNA_target_bh_edge_features, duplex_edge_features, 
                 reaction_rates):
        super().__init__()
        self.crna_edge_indices = crna_edge_indices
        self.ssDNA_target_bh_edge_indices = ssDNA_target_bh_edge_indices
        self.duplex_edge_indices = duplex_edge_indices
        self.crna_edge_features = crna_edge_features
        self.ssDNA_target_bh_edge_features = ssDNA_target_bh_edge_features
        self.duplex_edge_features = duplex_edge_features
        self.reaction_rates = reaction_rates
        self.num_samples = len(reaction_rates)
    def __len__(self):
        return self.num_samples
    def __getitem__(self, idx):
        c_edge_index_list = self.crna_edge_indices[idx]
        if not c_edge_index_list:
            c_edge_index = torch.empty((2,0), dtype=torch.long)
            n_c = 20  # fallback (e.g., 20-nt)
        else:
            c_edge_index = torch.tensor(c_edge_index_list, dtype=torch.long)
            n_c = _num_nodes_from_edge_index(c_edge_index_list, fallback=20)
        c_x = torch.zeros((n_c, 4), dtype=torch.float)  # placeholder
        c_edge_attr = torch.tensor(self.crna_edge_features[idx], dtype=torch.float)
        if c_edge_attr.dim() == 1:
            c_edge_attr = c_edge_attr.unsqueeze(1)   # [E,1]
        cRNA_data = Data(x=c_x, edge_index=c_edge_index, edge_attr=c_edge_attr)

        t_bh_edge_index_list = self.ssDNA_target_bh_edge_indices[idx]
        if not t_bh_edge_index_list:
            t_bh_edge_index = torch.empty((2,0), dtype=torch.long)
            n_t = 20
        else:
            t_bh_edge_index = torch.tensor(t_bh_edge_index_list, dtype=torch.long)
            n_t = _num_nodes_from_edge_index(t_bh_edge_index_list, fallback=20)
        t_bh_x = torch.zeros((n_t, 4), dtype=torch.float)
        t_bh_edge_attr = torch.tensor(self.ssDNA_target_bh_edge_features[idx], dtype=torch.float)
        if t_bh_edge_attr.dim() == 1:
            t_bh_edge_attr = t_bh_edge_attr.unsqueeze(1)   # [E,1]
        ssDNA_target_bh_data = Data(x=t_bh_x, edge_index=t_bh_edge_index, edge_attr=t_bh_edge_attr)
        
        d_edge_index_list = self.duplex_edge_indices[idx]
        if not d_edge_index_list:
            d_edge_index = torch.empty((2,0), dtype=torch.long)
            n_d = 40
        else:
            d_edge_index = torch.tensor(d_edge_index_list, dtype=torch.long)
            n_d = _num_nodes_from_edge_index(d_edge_index_list, fallback=40)
        d_x = torch.zeros((n_d, 4), dtype=torch.float)
        d_edge_attr = torch.tensor(self.duplex_edge_features[idx], dtype=torch.float)
        if d_edge_attr.dim() == 1:
            d_edge_attr = d_edge_attr.unsqueeze(1)   # [E,1]
        duplex_data = Data(x=d_x, edge_index=d_edge_index, edge_attr=d_edge_attr)
        
        y_val = torch.tensor([self.reaction_rates[idx]], dtype=torch.float)
        return cRNA_data, ssDNA_target_bh_data, duplex_data, y_val

def graph_collate(batch):
    from torch_geometric.data import Batch
    cRNA_list, ssDNA_target_bh_list, duplex_list, y_list = zip(*batch)
    batch_cRNA = Batch.from_data_list(cRNA_list)
    batch_ssDNA_target_bh = Batch.from_data_list(ssDNA_target_bh_list)
    batch_duplex = Batch.from_data_list(duplex_list)
    y = torch.tensor(y_list, dtype=torch.float).view(-1)
    return batch_cRNA, batch_ssDNA_target_bh, batch_duplex, y



########################################
# 3. CNN, GNN, MLP Branches (One hidden MLP)
########################################

class CNNBranch1(nn.Module):
    """
    CNN branch for (60,4).
    """
    def __init__(self, filters, kernel_size, dense_units):
        super().__init__()
        self.conv = nn.Conv1d(4, filters, kernel_size, padding=(kernel_size-1)//2)
        self.pool = nn.MaxPool1d(2)
        self.fc = nn.Linear(filters*30, dense_units)
    def forward(self, x):
        x = x.permute(0,2,1)  # (batch,4,30)
        x = F.relu(self.conv(x))
        x = self.pool(x)      # (batch,filters,30)
        x = x.flatten(start_dim=1)
        x = F.relu(self.fc(x))
        return x
        

class CNNBranch2(nn.Module):
    """
    CNN branch for (90,3).
    """
    def __init__(self, filters, kernel_size, dense_units):
        super().__init__()
        self.conv = nn.Conv1d(3, filters, kernel_size, padding=(kernel_size-1)//2)
        self.pool = nn.MaxPool1d(2)
        self.fc = nn.Linear(filters*45, dense_units)
    def forward(self, x):
        x = x.permute(0,2,1)  # (batch,3,45)
        x = F.relu(self.conv(x))
        x = self.pool(x)      # (batch,filters,45)
        x = x.flatten(start_dim=1)
        x = F.relu(self.fc(x))
        return x

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch_geometric.nn import NNConv, global_mean_pool, global_max_pool


class GNNBranch(nn.Module):
    def __init__(self, hidden_dim=32, edge_attr_dim=1):
        super().__init__()
        self.hidden_dim = hidden_dim

        def make_edge_mlp(edge_attr_dim, in_channels, out_channels):
            return nn.Sequential(
                nn.Linear(edge_attr_dim, hidden_dim),
                nn.ReLU(),
                nn.Linear(hidden_dim, in_channels * out_channels)
            )

        self.edge_mlp_c   = make_edge_mlp(edge_attr_dim, 4, hidden_dim)
        self.edge_mlp_c2  = make_edge_mlp(edge_attr_dim, hidden_dim, hidden_dim)
        self.conv1_c = NNConv(4, hidden_dim, self.edge_mlp_c, aggr='mean')
        self.conv2_c = NNConv(hidden_dim, hidden_dim, self.edge_mlp_c2, aggr='mean')

        self.edge_mlp_bh  = make_edge_mlp(edge_attr_dim, 4, hidden_dim)
        self.edge_mlp_bh2 = make_edge_mlp(edge_attr_dim, hidden_dim, hidden_dim)
        self.conv1_t_bh = NNConv(4, hidden_dim, self.edge_mlp_bh, aggr='mean')
        self.conv2_t_bh = NNConv(hidden_dim, hidden_dim, self.edge_mlp_bh2, aggr='mean')

        self.edge_mlp_d   = make_edge_mlp(edge_attr_dim, 4, hidden_dim)
        self.edge_mlp_d2  = make_edge_mlp(edge_attr_dim, hidden_dim, hidden_dim)
        self.conv1_d = NNConv(4, hidden_dim, self.edge_mlp_d, aggr='mean')
        self.conv2_d = NNConv(hidden_dim, hidden_dim, self.edge_mlp_d2, aggr='mean')

        self.mlp_merge = nn.Sequential(
            nn.Linear(hidden_dim * 3 * 2, hidden_dim),  # 3 graphs, mean+max
            nn.ReLU(),
            nn.Linear(hidden_dim, hidden_dim)
        )

    def pool(self, x, batch):
        return torch.cat([global_mean_pool(x, batch), global_max_pool(x, batch)], dim=1)

    def forward(self, crna_data, ssDNA_target_bh_data, duplex_data):
        def process(graph_data, conv1, conv2):
            x, edge_index, edge_attr, batch = graph_data.x, graph_data.edge_index, graph_data.edge_attr, graph_data.batch
            x = F.elu(conv1(x, edge_index, edge_attr))   # << correct order
            x = F.elu(conv2(x, edge_index, edge_attr))
            return self.pool(x, batch)

        x_c   = process(crna_data, self.conv1_c,   self.conv2_c)
        x_tbh = process(ssDNA_target_bh_data, self.conv1_t_bh, self.conv2_t_bh)
        x_d   = process(duplex_data, self.conv1_d, self.conv2_d)

        merged = torch.cat([x_c, x_tbh, x_d], dim=1)
        return self.mlp_merge(merged)


class MLPBranch120(nn.Module):
    """
    A single hidden layer MLP for 120D => output dimension mlp_dim.
    No separate "output" layer; just one fc + ReLU => final embedding.
    """
    def __init__(self, input_dim=120, hidden_dim=32):
        super().__init__()
        self.fc = nn.Linear(input_dim, hidden_dim)
    def forward(self, x):
        # x shape: (batch,120)
        x = F.relu(self.fc(x))  # (batch,hidden_dim)
        return x

########################################
# 4. Final Fusion Model
########################################

class CNN_GNN_MLP_Fusion(nn.Module):
    """
    End-to-end: 
      - CNNBranch1 => feat1
      - CNNBranch2 => feat2
      - GNNBranch  => feat_gnn
      - MLPBranch120 => feat_mlp
    Concat => dropout => final FC => 1
    """
    def __init__(self,
                 filters1, kernel_size1, dense_units1,  # CNN1
                 filters2, kernel_size2, dense_units2,  # CNN2
                 gnn_hidden_dim,
                 mlp_hidden_dim,  # single hidden dimension for MLP
                 final_fc_dim,
                 dropout_rate=0.0):  # new hyperparameter for dropout
        super().__init__()
        
        self.cnn_branch1 = CNNBranch1(filters1, kernel_size1, dense_units1)
        self.cnn_branch2 = CNNBranch2(filters2, kernel_size2, dense_units2)
        self.gnn_branch = GNNBranch(hidden_dim=gnn_hidden_dim)
        self.mlp_branch = MLPBranch120(input_dim=120, hidden_dim=mlp_hidden_dim)
        
        # total dimension = (dense_units1 + dense_units2 + gnn_hidden_dim + mlp_hidden_dim)
        total_dim = dense_units1 + dense_units2 + gnn_hidden_dim + mlp_hidden_dim
        
        self.dropout_rate = dropout_rate
        self.hidden = nn.Linear(total_dim, final_fc_dim)
        self.out = nn.Linear(final_fc_dim, 1)

    
    def forward(self, crna_data, ssDNA_target_bh_data, duplex_data, x1, x2, x3):
        feat1 = self.cnn_branch1(x1)                   # (batch, dense_units1)
        feat2 = self.cnn_branch2(x2)                   # (batch, dense_units2)
        feat_gnn = self.gnn_branch(crna_data, ssDNA_target_bh_data, duplex_data)  # (batch, gnn_hidden_dim)
        feat_mlp = self.mlp_branch(x3)                 # (batch, mlp_hidden_dim)
        
        merged = torch.cat([feat1, feat2, feat_gnn, feat_mlp], dim=1)
        # Apply dropout on the concatenated features
        merged = F.dropout(merged, p=self.dropout_rate, training=self.training)
        x = F.relu(self.hidden(merged))   # (batch, final_fc_dim)
        out = self.out(x)                 # (batch, 1)
        return out.view(-1)


########################################
# 5. Hybrid Dataset
########################################

class HybridDataset(Dataset):
    def __init__(self, graph_dataset, X1, X2, X3, reaction_rates):
        super().__init__()
        self.graph_dataset = graph_dataset
        self.X1 = X1
        self.X2 = X2
        self.X3 = X3
        self.reaction_rates = reaction_rates
        self.num_samples = len(reaction_rates)
        
        assert len(graph_dataset) == self.num_samples
        assert len(X1) == self.num_samples
        assert len(X2) == self.num_samples
        assert len(X3) == self.num_samples
    
    def __len__(self):
        return self.num_samples
    
    def __getitem__(self, idx):
        crna_data, ssDNA_target_bh_data, duplex_data, y_val = self.graph_dataset[idx]
        x1 = torch.tensor(self.X1[idx], dtype=torch.float)  
        x2 = torch.tensor(self.X2[idx], dtype=torch.float)  
        x3 = torch.tensor(self.X3[idx], dtype=torch.float)  
        return crna_data, ssDNA_target_bh_data, duplex_data, x1, x2, x3, y_val

def hybrid_collate(batch):
    from torch_geometric.data import Batch
    cRNA_list, ssDNA_target_bh_list, duplex_list, x1_list, x2_list, x3_list, y_list = zip(*batch)
    batch_cRNA = Batch.from_data_list(cRNA_list)
    batch_ssDNA_target_bh = Batch.from_data_list(ssDNA_target_bh_list)
    batch_duplex = Batch.from_data_list(duplex_list)
    x1 = torch.stack(x1_list, dim=0)
    x2 = torch.stack(x2_list, dim=0)
    x3 = torch.stack(x3_list, dim=0)
    y = torch.tensor(y_list, dtype=torch.float).view(-1)
    return batch_cRNA, batch_ssDNA_target_bh, batch_duplex, x1, x2, x3, y

########################################
# 6. Objective Function (no pretraining)
########################################

import copy

def objective(trial):
    # CNN1
    filters1 = trial.suggest_int("filters1", 16, 128, step=16)
    kernel_size1 = trial.suggest_categorical("kernel_size1", [3, 5, 7])
    dense_units1 = trial.suggest_int("dense_units1", 16, 256, step=16)
    
    # CNN2
    filters2 = trial.suggest_int("filters2", 16, 128, step=16)
    kernel_size2 = trial.suggest_categorical("kernel_size2", [3,5,7])
    dense_units2 = trial.suggest_int("dense_units2", 16, 256, step=16)
    
    # GNN
    gnn_hidden_dim = trial.suggest_int("gnn_hidden_dim", 16, 64, step=16)
    
    # MLP: single hidden layer => "mlp_hidden_dim"
    mlp_hidden_dim = trial.suggest_int("mlp_hidden_dim", 16, 256, step=16)
    
    # final aggregator
    final_fc_dim = trial.suggest_int("final_fc_dim", 16, 256, step=16)

    # Dropout hyperparameter (0.2 to 0.7)
    dropout_rate = trial.suggest_float("dropout_rate", 0.0, 0.5)
    
    # Learning rate and batch size
    lr = trial.suggest_loguniform("lr", 1e-4, 1e-2)
    batch_size = trial.suggest_categorical("batch_size", [8,16,32])
    epochs = 1000
    patience = 10
    
    loss_fn = nn.MSELoss()
    
    # load data
    X1_HTCas9 = load_branch1_data_HTCas9()
    X1_HT11 = load_branch1_data_HT11()
    X1_mmc4 = load_branch1_data_mmc4()
    X1_mmc5 = load_branch1_data_mmc5()
    X1_change_seq = load_branch1_data_change_seq()
    X1_tiger = load_branch1_data_tiger()
    X1_iMeta = load_branch1_data_iMeta()
    X1_HTCas9 = np.asarray(X1_HTCas9)
    X1_HT11   = np.asarray(X1_HT11)
    X1_mmc4   = np.asarray(X1_mmc4)
    X1_mmc5   = np.asarray(X1_mmc5)
    X1_change_seq   = np.asarray(X1_change_seq)
    X1_tiger   = np.asarray(X1_tiger)
    X1_iMeta   = np.asarray(X1_iMeta)
    X1 = np.concatenate([X1_HTCas9, X1_HT11, X1_mmc4, X1_mmc5, X1_change_seq, X1_tiger, X1_iMeta], axis=0) 
    
    X2_HTCas9 = load_branch2_data_HTCas9()
    X2_HT11 = load_branch2_data_HT11()
    X2_mmc4 = load_branch2_data_mmc4()
    X2_mmc5 = load_branch2_data_mmc5()
    X2_change_seq = load_branch2_data_change_seq()
    X2_tiger = load_branch2_data_tiger()
    X2_iMeta = load_branch2_data_iMeta()
    X2_HTCas9 = np.asarray(X2_HTCas9)
    X2_HT11   = np.asarray(X2_HT11)
    X2_mmc4 = np.asarray(X2_mmc4)
    X2_mmc5   = np.asarray(X2_mmc5)
    X2_change_seq   = np.asarray(X2_change_seq)
    X2_tiger   = np.asarray(X2_tiger)
    X2_iMeta   = np.asarray(X2_iMeta)
    X2 = np.concatenate([X2_HTCas9, X2_HT11, X2_mmc4, X2_mmc5, X2_change_seq, X2_tiger, X2_iMeta], axis=0) 

    X3_HTCas9 = load_branch3_data_HTCas9()
    X3_HT11 = load_branch3_data_HT11()
    X3_mmc4 = load_branch3_data_mmc4()
    X3_mmc5 = load_branch3_data_mmc5()
    X3_change_seq = load_branch3_data_change_seq()
    X3_tiger = load_branch3_data_tiger()
    X3_iMeta = load_branch3_data_iMeta()
    X3_HTCas9 = np.asarray(X3_HTCas9)
    X3_HT11   = np.asarray(X3_HT11)
    X3_mmc4 = np.asarray(X3_mmc4)
    X3_mmc5   = np.asarray(X3_mmc5)
    X3_change_seq   = np.asarray(X3_change_seq)
    X3_tiger   = np.asarray(X3_tiger)
    X3_iMeta   = np.asarray(X3_iMeta)
    X3 = np.concatenate([X3_HTCas9, X3_HT11, X3_mmc4, X3_mmc5, X3_change_seq, X3_tiger, X3_iMeta], axis=0) 

    rates_HTCas9 = load_reaction_rates_HTCas9()
    rates_HT11 = load_reaction_rates_HT11()
    rates_mmc4 = load_reaction_rates_mmc4()
    rates_mmc5 = load_reaction_rates_mmc5()
    rates_change_seq = load_reaction_rates_change_seq()
    rates_tiger = load_reaction_rates_tiger()
    rates_iMeta = load_reaction_rates_iMeta()
    rates_HTCas9 = np.asarray(rates_HTCas9)
    rates_HT11   = np.asarray(rates_HT11)
    rates_mmc4 = np.asarray(rates_mmc4)
    rates_mmc5   = np.asarray(rates_mmc5)
    rates_change_seq   = np.asarray(rates_change_seq)
    rates_tiger   = np.asarray(rates_tiger)
    rates_iMeta   = np.asarray(rates_iMeta)
    rates = np.concatenate([rates_HTCas9, rates_HT11, rates_mmc4, rates_mmc5, rates_change_seq, rates_tiger, rates_iMeta], axis=0) 
    
    # Load both sources
    data_HTCas9 = load_dna_reaction_data_HTCas9()
    data_HT11   = load_dna_reaction_data_HT11()
    data_mmc4   = load_rna_reaction_data_mmc4()
    data_mmc5   = load_rna_reaction_data_mmc5()
    data_change_seq  = load_dna_reaction_data_change_seq()
    data_tiger  = load_rna_reaction_data_tiger()
    data_iMeta   = load_dna_reaction_data_iMeta()
    
    # Combine
    combined = combine_dna_graph_sets(data_HTCas9, data_HT11, data_mmc4, data_mmc5, data_change_seq, data_tiger, data_iMeta)
    
    # Unpack if needed
    (crna_edge_indices, ssDNA_target_bh_edge_indices, duplex_edge_indices,
     crna_edge_features, ssDNA_target_bh_edge_features, duplex_edge_features,
     reaction_rates) = combined
    
    
    graph_dataset = CRNADuplexSubstructuresDataset(
        crna_edge_indices, ssDNA_target_bh_edge_indices, duplex_edge_indices, 
        crna_edge_features, ssDNA_target_bh_edge_features, duplex_edge_features, 
        rates
    )
    hybrid_dataset = HybridDataset(graph_dataset, X1, X2, X3, rates)
    
    # Splitting the HTCas9 dataset
    np.random.seed(42)
    full_indices_HTCas9 = np.arange(len(rates_HTCas9))
    selected_indices_HTCas9 = np.random.choice(len(full_indices_HTCas9), size=10682, replace=False)
    unseen_indices_HTCas9 = np.setdiff1d(full_indices_HTCas9, selected_indices_HTCas9)
    train_idx_HTCas9, val_idx_HTCas9 = train_test_split(selected_indices_HTCas9, test_size=0.2, random_state=42)
    
    train_set_HTCas9 = Subset(hybrid_dataset, train_idx_HTCas9)
    val_set_HTCas9 = Subset(hybrid_dataset, val_idx_HTCas9)

    # Splitting the HT11 dataset
    np.random.seed(42)
    full_indices_HT11 = np.arange(len(rates_HT11))
    selected_indices_HT11 = np.random.choice(len(full_indices_HT11), size=9092, replace=False)
    unseen_indices_HT11 = np.setdiff1d(full_indices_HT11, selected_indices_HT11)
    # how many samples belong to HTCas9 at the front of hybrid_dataset
    offset = len(rates_HTCas9) 
    train_idx_HT11, val_idx_HT11 = train_test_split(selected_indices_HT11, test_size=0.2, random_state=42)
    # shift HT11 indices into the global index space of hybrid_dataset
    train_idx_HT11_global = (train_idx_HT11 + offset).tolist()
    val_idx_HT11_global   = (val_idx_HT11   + offset).tolist()
    # now these Subsets will point to the HT11 region of hybrid_dataset
    train_set_HT11 = Subset(hybrid_dataset, train_idx_HT11_global)
    val_set_HT11   = Subset(hybrid_dataset, val_idx_HT11_global)

     # Splitting the mmc4 dataset
    np.random.seed(42)
    full_indices_mmc4 = np.arange(len(rates_mmc4))
    selected_indices_mmc4 = np.random.choice(len(full_indices_mmc4), size=1742, replace=False)
    unseen_indices_mmc4 = np.setdiff1d(full_indices_mmc4, selected_indices_mmc4)
    # how many samples belong to HTCas9 at the front of hybrid_dataset
    offset = len(rates_HTCas9) + len(rates_HT11)
    train_idx_mmc4, val_idx_mmc4 = train_test_split(selected_indices_mmc4, test_size=0.2, random_state=42)
    # shift mmc4 indices into the global index space of hybrid_dataset
    train_idx_mmc4_global = (train_idx_mmc4 + offset).tolist()
    val_idx_mmc4_global   = (val_idx_mmc4   + offset).tolist()
    # now these Subsets will point to the mmc4 region of hybrid_dataset
    train_set_mmc4 = Subset(hybrid_dataset, train_idx_mmc4_global)
    val_set_mmc4   = Subset(hybrid_dataset, val_idx_mmc4_global)

    # Splitting the mmc5 dataset
    np.random.seed(42)
    full_indices_mmc5 = np.arange(len(rates_mmc5))
    selected_indices_mmc5 = np.random.choice(len(full_indices_mmc5), size=1835, replace=False)
    unseen_indices_mmc5 = np.setdiff1d(full_indices_mmc5, selected_indices_mmc5)
    # how many samples belong to HTCas9 at the front of hybrid_dataset
    offset = len(rates_HTCas9) + len(rates_HT11) + len(rates_mmc4)
    train_idx_mmc5, val_idx_mmc5 = train_test_split(selected_indices_mmc5, test_size=0.2, random_state=42)
    # shift mmc5 indices into the global index space of hybrid_dataset
    train_idx_mmc5_global = (train_idx_mmc5 + offset).tolist()
    val_idx_mmc5_global   = (val_idx_mmc5   + offset).tolist()
    # now these Subsets will point to the mmc5 region of hybrid_dataset
    train_set_mmc5 = Subset(hybrid_dataset, train_idx_mmc5_global)
    val_set_mmc5   = Subset(hybrid_dataset, val_idx_mmc5_global)

    # Splitting the change_seq dataset
    np.random.seed(42)
    full_indices_change_seq = np.arange(len(rates_change_seq))
    selected_indices_change_seq = np.random.choice(len(full_indices_change_seq), size=8000, replace=False)
    unseen_indices_change_seq = np.setdiff1d(full_indices_change_seq, selected_indices_change_seq)
    # how many samples belong to HTCas9 at the front of hybrid_dataset
    offset = len(rates_HTCas9) + len(rates_HT11) + len(rates_mmc4) + len(rates_mmc5)
    train_idx_change_seq, val_idx_change_seq = train_test_split(selected_indices_change_seq, test_size=0.2, random_state=42)
    # shift change_seq indices into the global index space of hybrid_dataset
    train_idx_change_seq_global = (train_idx_change_seq + offset).tolist()
    val_idx_change_seq_global   = (val_idx_change_seq   + offset).tolist()
    # now these Subsets will point to the change_seq region of hybrid_dataset
    train_set_change_seq = Subset(hybrid_dataset, train_idx_change_seq_global)
    val_set_change_seq   = Subset(hybrid_dataset, val_idx_change_seq_global)

    # Splitting the tiger dataset
    np.random.seed(42)
    full_indices_tiger = np.arange(len(rates_tiger))
    selected_indices_tiger = np.random.choice(len(full_indices_tiger), size=8000, replace=False)
    unseen_indices_tiger = np.setdiff1d(full_indices_tiger, selected_indices_tiger)
    # how many samples belong to HTCas9 at the front of hybrid_dataset
    offset = len(rates_HTCas9) + len(rates_HT11) + len(rates_mmc4) + len(rates_mmc5) + len(rates_change_seq)
    train_idx_tiger, val_idx_tiger = train_test_split(selected_indices_tiger, test_size=0.2, random_state=42)
    # shift tiger indices into the global index space of hybrid_dataset
    train_idx_tiger_global = (train_idx_tiger + offset).tolist()
    val_idx_tiger_global   = (val_idx_tiger   + offset).tolist()
    # now these Subsets will point to the tiger region of hybrid_dataset
    train_set_tiger = Subset(hybrid_dataset, train_idx_tiger_global)
    val_set_tiger   = Subset(hybrid_dataset, val_idx_tiger_global)

    # Splitting the iMeta dataset
    np.random.seed(42)
    full_indices_iMeta = np.arange(len(rates_iMeta))
    selected_indices_iMeta = np.random.choice(len(full_indices_iMeta), size=5776, replace=False)
    unseen_indices_iMeta = np.setdiff1d(full_indices_iMeta, selected_indices_iMeta)
    # how many samples belong to HTCas9 at the front of hybrid_dataset
    offset = len(rates_HTCas9) + len(rates_HT11) + len(rates_mmc4) + len(rates_mmc5) + len(rates_change_seq) + len(rates_tiger)
    train_idx_iMeta, val_idx_iMeta = train_test_split(selected_indices_iMeta, test_size=0.2, random_state=42)
    # shift iMeta indices into the global index space of hybrid_dataset
    train_idx_iMeta_global = (train_idx_iMeta + offset).tolist()
    val_idx_iMeta_global   = (val_idx_iMeta   + offset).tolist()
    # now these Subsets will point to the iMeta region of hybrid_dataset
    train_set_iMeta = Subset(hybrid_dataset, train_idx_iMeta_global)
    val_set_iMeta   = Subset(hybrid_dataset, val_idx_iMeta_global)



    from torch.utils.data import ConcatDataset, DataLoader

    # combine the subsets (datasets)
    train_set = ConcatDataset([train_set_HTCas9, train_set_HT11, train_set_mmc4, train_set_mmc5, train_set_change_seq, train_set_tiger, train_set_iMeta])
    val_set   = ConcatDataset([val_set_HTCas9,   val_set_HT11,   val_set_mmc4,   val_set_mmc5, val_set_change_seq, val_set_tiger, val_set_iMeta])
    
    # build new loaders
    train_loader = DataLoader(train_set, batch_size=batch_size, shuffle=True, collate_fn=hybrid_collate)
    val_loader   = DataLoader(val_set,   batch_size=batch_size, shuffle=False, collate_fn=hybrid_collate)
    

    # Build the model
    model = CNN_GNN_MLP_Fusion(
        filters1, kernel_size1, dense_units1,
        filters2, kernel_size2, dense_units2,
        gnn_hidden_dim,
        mlp_hidden_dim,
        final_fc_dim,
        dropout_rate=dropout_rate
    )
    optimizer = optim.Adam(model.parameters(), lr=lr, weight_decay=1e-4)
    
    best_val_loss = float('inf')
    counter = 0
    best_model_state = None  # To store the best model's state
    
    for epoch in range(epochs):
        # Training phase
        model.train()
        total_loss = 0.0
        for crna_b, ssDNA_target_bh_b, duplex_b, x1_b, x2_b, x3_b, y_b in train_loader:
            pred = model(crna_b, ssDNA_target_bh_b, duplex_b, x1_b, x2_b, x3_b)
            loss = loss_fn(pred, y_b.view(-1))
            optimizer.zero_grad()
            loss.backward()
            optimizer.step()
            total_loss += loss.item()
        avg_train_loss = total_loss / len(train_loader)
        
        # Validation phase
        model.eval()
        total_val = 0.0
        with torch.no_grad():
            for crna_b, ssDNA_target_bh_b, duplex_b, x1_b, x2_b, x3_b, y_b in val_loader:
                valpred = model(crna_b, ssDNA_target_bh_b, duplex_b, x1_b, x2_b, x3_b)
                vloss = loss_fn(valpred, y_b.view(-1))
                total_val += vloss.item()
        avg_val_loss = total_val / len(val_loader)
        
        trial.report(avg_val_loss, epoch)
        
        # Check for improvement
        if avg_val_loss < best_val_loss:
            best_val_loss = avg_val_loss
            best_model_state = copy.deepcopy(model)
            counter = 0
        else:
            counter += 1
        
        # Early stopping if no improvement for 'patience' epochs
        if counter >= patience:
            print(f"Early stopping at epoch {epoch+1}")
            break
        
        print(f"Epoch {epoch+1}/{epochs} | Train: {avg_train_loss:.4f} | Val: {avg_val_loss:.4f}")
    
    # Load the best model before saving
    if best_model_state is not None:
        model = best_model_state
    # Save the entire model instead of the state dictionary
    torch.save(model, f"best_generalist_FULL.pt")

    # how many samples at the front of hybrid_dataset
    offset_1 = len(rates_HTCas9)
    offset_2 = len(rates_HTCas9) + len(rates_HT11)
    offset_3 = len(rates_HTCas9) + len(rates_HT11) + len(rates_mmc4)
    offset_4 = len(rates_HTCas9) + len(rates_HT11) + len(rates_mmc4) + len(rates_mmc5)
    offset_5 = len(rates_HTCas9) + len(rates_HT11) + len(rates_mmc4) + len(rates_mmc5) + len(rates_change_seq)
    offset_6 = len(rates_HTCas9) + len(rates_HT11) + len(rates_mmc4) + len(rates_mmc5) + len(rates_change_seq) + len(rates_tiger)
    # shift indices into the global index space of hybrid_dataset
    unseen_indices_HT11_global = (unseen_indices_HT11 + offset_1).tolist()
    unseen_indices_mmc4_global = (unseen_indices_mmc4 + offset_2).tolist()
    unseen_indices_mmc5_global = (unseen_indices_mmc5 + offset_3).tolist()
    unseen_indices_change_seq_global = (unseen_indices_change_seq + offset_4).tolist()
    unseen_indices_tiger_global = (unseen_indices_tiger + offset_5).tolist()
    unseen_indices_iMeta_global = (unseen_indices_iMeta + offset_6).tolist()
    
    # Evaluate on 10 subsamples from unseen
    unseen_set_HTCas9 = Subset(hybrid_dataset, unseen_indices_HTCas9)
    unseen_set_HT11 = Subset(hybrid_dataset, unseen_indices_HT11_global)
    unseen_set_mmc4 = Subset(hybrid_dataset, unseen_indices_mmc4_global)
    unseen_set_mmc5 = Subset(hybrid_dataset, unseen_indices_mmc5_global)
    unseen_set_change_seq = Subset(hybrid_dataset, unseen_indices_change_seq_global)
    unseen_set_tiger = Subset(hybrid_dataset, unseen_indices_tiger_global)
    unseen_set_iMeta = Subset(hybrid_dataset, unseen_indices_iMeta_global)
        
    model.eval()


    # Compute unseen Spearman for HTCas9 dataset
    spearman_list_HTCas9 = []
    for i in range(10):
        if len(unseen_set_HTCas9) < 1000:
            break
        idx_trial = np.random.choice(len(unseen_set_HTCas9), size=1000, replace=False)
        trial_subset = Subset(unseen_set_HTCas9, idx_trial)
        trial_loader = DataLoader(trial_subset, batch_size=1, shuffle=False, collate_fn=hybrid_collate)
        
        preds_trial = []
        labels_trial = []
        with torch.no_grad():
            for c_b, tbh_b, d_b, xx1_b, xx2_b, xx3_b, yy_b in trial_loader:
                p = model(c_b, tbh_b, d_b, xx1_b, xx2_b, xx3_b)
                preds_trial.append(p.item())
                labels_trial.append(yy_b.item())
        sp_corr, _ = spearmanr(labels_trial, preds_trial)
        spearman_list_HTCas9.append(sp_corr)
        print(f"Subsample {i+1}: Spearman = {sp_corr:.4f}")
    avg_spearman = np.mean(spearman_list_HTCas9) if spearman_list_HTCas9 else None
    print(f"Average Spearman over 10 subsamples in HTCas9 dataset: {avg_spearman}")



    # Compute unseen Spearman for HT11 dataset
    spearman_list_HT11 = []
    for i in range(10):
        if len(unseen_set_HT11) < 1000:
            break
        idx_trial = np.random.choice(len(unseen_set_HT11), size=1000, replace=False)
        trial_subset = Subset(unseen_set_HT11, idx_trial)
        trial_loader = DataLoader(trial_subset, batch_size=1, shuffle=False, collate_fn=hybrid_collate)
        
        preds_trial = []
        labels_trial = []
        with torch.no_grad():
            for c_b, tbh_b, d_b, xx1_b, xx2_b, xx3_b, yy_b in trial_loader:
                p = model(c_b, tbh_b, d_b, xx1_b, xx2_b, xx3_b)
                preds_trial.append(p.item())
                labels_trial.append(yy_b.item())
        sp_corr, _ = spearmanr(labels_trial, preds_trial)
        spearman_list_HT11.append(sp_corr)
        print(f"Subsample {i+1}: Spearman = {sp_corr:.4f}")
    avg_spearman = np.mean(spearman_list_HT11) if spearman_list_HT11 else None
    print(f"Average Spearman over 10 subsamples in HT11 dataset: {avg_spearman}")


    # Compute unseen Spearman for mmc4 dataset
    spearman_list_mmc4 = []
    for i in range(10):
        if len(unseen_set_mmc4) < 300:
            break
        idx_trial = np.random.choice(len(unseen_set_mmc4), size=300, replace=False)
        trial_subset = Subset(unseen_set_mmc4, idx_trial)
        trial_loader = DataLoader(trial_subset, batch_size=1, shuffle=False, collate_fn=hybrid_collate)
        
        preds_trial = []
        labels_trial = []
        with torch.no_grad():
            for c_b, tbh_b, d_b, xx1_b, xx2_b, xx3_b, yy_b in trial_loader:
                p = model(c_b, tbh_b, d_b, xx1_b, xx2_b, xx3_b)
                preds_trial.append(p.item())
                labels_trial.append(yy_b.item())
        sp_corr, _ = spearmanr(labels_trial, preds_trial)
        spearman_list_mmc4.append(sp_corr)
        print(f"Subsample {i+1}: Spearman = {sp_corr:.4f}")
    avg_spearman = np.mean(spearman_list_mmc4) if spearman_list_mmc4 else None
    print(f"Average Spearman over 10 subsamples in mmc4 dataset: {avg_spearman}")


    # Compute unseen Spearman for mmc5 dataset
    spearman_list_mmc5 = []
    for i in range(10):
        if len(unseen_set_mmc5) < 300:
            break
        idx_trial = np.random.choice(len(unseen_set_mmc5), size=300, replace=False)
        trial_subset = Subset(unseen_set_mmc5, idx_trial)
        trial_loader = DataLoader(trial_subset, batch_size=1, shuffle=False, collate_fn=hybrid_collate)
        
        preds_trial = []
        labels_trial = []
        with torch.no_grad():
            for c_b, tbh_b, d_b, xx1_b, xx2_b, xx3_b, yy_b in trial_loader:
                p = model(c_b, tbh_b, d_b, xx1_b, xx2_b, xx3_b)
                preds_trial.append(p.item())
                labels_trial.append(yy_b.item())
        sp_corr, _ = spearmanr(labels_trial, preds_trial)
        spearman_list_mmc5.append(sp_corr)
        print(f"Subsample {i+1}: Spearman = {sp_corr:.4f}")
    avg_spearman = np.mean(spearman_list_mmc5) if spearman_list_mmc5 else None
    print(f"Average Spearman over 10 subsamples in mmc5 dataset: {avg_spearman}")

    # Compute unseen Spearman for change_seq dataset
    spearman_list_change_seq = []
    for i in range(10):
        if len(unseen_set_change_seq) < 1000:
            break
        idx_trial = np.random.choice(len(unseen_set_change_seq), size=1000, replace=False)
        trial_subset = Subset(unseen_set_change_seq, idx_trial)
        trial_loader = DataLoader(trial_subset, batch_size=1, shuffle=False, collate_fn=hybrid_collate)
        
        preds_trial = []
        labels_trial = []
        with torch.no_grad():
            for c_b, tbh_b, d_b, xx1_b, xx2_b, xx3_b, yy_b in trial_loader:
                p = model(c_b, tbh_b, d_b, xx1_b, xx2_b, xx3_b)
                preds_trial.append(p.item())
                labels_trial.append(yy_b.item())
        sp_corr, _ = spearmanr(labels_trial, preds_trial)
        spearman_list_change_seq.append(sp_corr)
        print(f"Subsample {i+1}: Spearman = {sp_corr:.4f}")
    avg_spearman = np.mean(spearman_list_change_seq) if spearman_list_change_seq else None
    print(f"Average Spearman over 10 subsamples in change_seq dataset: {avg_spearman}")

    # Compute unseen Spearman for tiger dataset
    spearman_list_tiger = []
    for i in range(10):
        if len(unseen_set_tiger) < 1000:
            break
        idx_trial = np.random.choice(len(unseen_set_tiger), size=1000, replace=False)
        trial_subset = Subset(unseen_set_tiger, idx_trial)
        trial_loader = DataLoader(trial_subset, batch_size=1, shuffle=False, collate_fn=hybrid_collate)
        
        preds_trial = []
        labels_trial = []
        with torch.no_grad():
            for c_b, tbh_b, d_b, xx1_b, xx2_b, xx3_b, yy_b in trial_loader:
                p = model(c_b, tbh_b, d_b, xx1_b, xx2_b, xx3_b)
                preds_trial.append(p.item())
                labels_trial.append(yy_b.item())
        sp_corr, _ = spearmanr(labels_trial, preds_trial)
        spearman_list_tiger.append(sp_corr)
        print(f"Subsample {i+1}: Spearman = {sp_corr:.4f}")
    avg_spearman = np.mean(spearman_list_tiger) if spearman_list_tiger else None   
    print(f"Average Spearman over 10 subsamples in tiger dataset: {avg_spearman}")

    # Compute unseen Spearman for iMeta dataset
    spearman_list_iMeta = []
    for i in range(10):
        if len(unseen_set_iMeta) < 1000:
            break
        idx_trial = np.random.choice(len(unseen_set_iMeta), size=1000, replace=False)
        trial_subset = Subset(unseen_set_iMeta, idx_trial)
        trial_loader = DataLoader(trial_subset, batch_size=1, shuffle=False, collate_fn=hybrid_collate)
        
        preds_trial = []
        labels_trial = []
        with torch.no_grad():
            for c_b, tbh_b, d_b, xx1_b, xx2_b, xx3_b, yy_b in trial_loader:
                p = model(c_b, tbh_b, d_b, xx1_b, xx2_b, xx3_b)
                preds_trial.append(p.item())
                labels_trial.append(yy_b.item())
        sp_corr, _ = spearmanr(labels_trial, preds_trial)
        spearman_list_iMeta.append(sp_corr)
        print(f"Subsample {i+1}: Spearman = {sp_corr:.4f}")
    avg_spearman = np.mean(spearman_list_iMeta) if spearman_list_iMeta else None   
    print(f"Average Spearman over 10 subsamples in iMeta dataset: {avg_spearman}")

    
    return best_val_loss

########################################
# 7. Main Pipeline
########################################

def main_pipeline():

    study = optuna.create_study(direction="minimize")
    study.optimize(objective, n_trials=10)
    
    print("Best hyperparameters:", study.best_trial.params)
    best_trial_number = study.best_trial.number
    best_params = study.best_trial.params
    
    # Rebuild for final 100-trial evaluation by loading the entire saved model
    model = torch.load(f"best_generalist_FULL.pt", weights_only=False)
    model.eval()
    
    # final hold-out evaluation
    # load data

    X1_HTCas9 = load_branch1_data_HTCas9()
    X1_HT11 = load_branch1_data_HT11()
    X1_mmc4 = load_branch1_data_mmc4()
    X1_mmc5 = load_branch1_data_mmc5()
    X1_change_seq = load_branch1_data_change_seq()
    X1_tiger = load_branch1_data_tiger()
    X1_iMeta = load_branch1_data_iMeta()
    X1_HTCas9 = np.asarray(X1_HTCas9)
    X1_HT11   = np.asarray(X1_HT11)
    X1_mmc4 = np.asarray(X1_mmc4)
    X1_mmc5   = np.asarray(X1_mmc5)
    X1_change_seq   = np.asarray(X1_change_seq)
    X1_tiger   = np.asarray(X1_tiger)
    X1_iMeta   = np.asarray(X1_iMeta)
    X1 = np.concatenate([X1_HTCas9, X1_HT11, X1_mmc4, X1_mmc5, X1_change_seq, X1_tiger, X1_iMeta], axis=0) 
    
    X2_HTCas9 = load_branch2_data_HTCas9()
    X2_HT11 = load_branch2_data_HT11()
    X2_mmc4 = load_branch2_data_mmc4()
    X2_mmc5 = load_branch2_data_mmc5()
    X2_change_seq = load_branch2_data_change_seq()
    X2_tiger = load_branch2_data_tiger()
    X2_iMeta = load_branch2_data_iMeta()
    X2_HTCas9 = np.asarray(X2_HTCas9)
    X2_HT11   = np.asarray(X2_HT11)
    X2_mmc4 = np.asarray(X2_mmc4)
    X2_mmc5   = np.asarray(X2_mmc5)
    X2_change_seq   = np.asarray(X2_change_seq)
    X2_tiger   = np.asarray(X2_tiger)
    X2_iMeta   = np.asarray(X2_iMeta)
    X2 = np.concatenate([X2_HTCas9, X2_HT11, X2_mmc4, X2_mmc5, X2_change_seq, X2_tiger, X2_iMeta], axis=0) 

    X3_HTCas9 = load_branch3_data_HTCas9()
    X3_HT11 = load_branch3_data_HT11()
    X3_mmc4 = load_branch3_data_mmc4()
    X3_mmc5 = load_branch3_data_mmc5()
    X3_change_seq = load_branch3_data_change_seq()
    X3_tiger = load_branch3_data_tiger()
    X3_iMeta = load_branch3_data_iMeta()
    X3_HTCas9 = np.asarray(X3_HTCas9)
    X3_HT11   = np.asarray(X3_HT11)
    X3_mmc4 = np.asarray(X3_mmc4)
    X3_mmc5   = np.asarray(X3_mmc5)
    X3_change_seq   = np.asarray(X3_change_seq)
    X3_tiger   = np.asarray(X3_tiger)
    X3_iMeta   = np.asarray(X3_iMeta)
    X3 = np.concatenate([X3_HTCas9, X3_HT11, X3_mmc4, X3_mmc5, X3_change_seq, X3_tiger, X3_iMeta], axis=0) 

    rates_HTCas9 = load_reaction_rates_HTCas9()
    rates_HT11 = load_reaction_rates_HT11()
    rates_mmc4 = load_reaction_rates_mmc4()
    rates_mmc5 = load_reaction_rates_mmc5()
    rates_change_seq = load_reaction_rates_change_seq()
    rates_tiger = load_reaction_rates_tiger()
    rates_iMeta = load_reaction_rates_iMeta()
    rates_HTCas9 = np.asarray(rates_HTCas9)
    rates_HT11   = np.asarray(rates_HT11)
    rates_mmc4 = np.asarray(rates_mmc4)
    rates_mmc5   = np.asarray(rates_mmc5)
    rates_change_seq   = np.asarray(rates_change_seq)
    rates_tiger   = np.asarray(rates_tiger)
    rates_iMeta   = np.asarray(rates_iMeta)
    rates = np.concatenate([rates_HTCas9, rates_HT11, rates_mmc4, rates_mmc5, rates_change_seq, rates_tiger, rates_iMeta], axis=0) 
    
    # Load both sources
    data_HTCas9 = load_dna_reaction_data_HTCas9()
    data_HT11   = load_dna_reaction_data_HT11()
    data_mmc4   = load_rna_reaction_data_mmc4()
    data_mmc5   = load_rna_reaction_data_mmc5()
    data_change_seq   = load_dna_reaction_data_change_seq()
    data_tiger   = load_rna_reaction_data_tiger()
    data_iMeta   = load_dna_reaction_data_iMeta()
    
    # Combine
    combined = combine_dna_graph_sets(data_HTCas9, data_HT11, data_mmc4, data_mmc5, data_change_seq, data_tiger, data_iMeta)
    
    # Unpack if needed
    (crna_edge_indices, ssDNA_target_bh_edge_indices, duplex_edge_indices,
     crna_edge_features, ssDNA_target_bh_edge_features, duplex_edge_features,
     reaction_rates) = combined
    
    graph_dataset = CRNADuplexSubstructuresDataset(
        crna_edge_indices, ssDNA_target_bh_edge_indices, duplex_edge_indices, 
        crna_edge_features, ssDNA_target_bh_edge_features, duplex_edge_features, 
        rates
    )
    hybrid_dataset = HybridDataset(graph_dataset, X1, X2, X3, rates)

    # Get unseen HTCas9 dataset
    np.random.seed(42)
    full_indices_HTCas9 = np.arange(len(rates_HTCas9))
    selected_indices_HTCas9 = np.random.choice(len(full_indices_HTCas9), size=10682, replace=False)
    unseen_indices_HTCas9 = np.setdiff1d(full_indices_HTCas9, selected_indices_HTCas9)

    # Splitting the HT11 dataset
    np.random.seed(42)
    full_indices_HT11 = np.arange(len(rates_HT11))
    selected_indices_HT11 = np.random.choice(len(full_indices_HT11), size=9092, replace=False)
    unseen_indices_HT11 = np.setdiff1d(full_indices_HT11, selected_indices_HT11)
    # how many samples belong to HTCas9 at the front of hybrid_dataset
    offset = len(rates_HTCas9)
    # shift HT11 indices into the global index space of hybrid_dataset
    unseen_indices_HT11_global = (unseen_indices_HT11 + offset).tolist()

    # Splitting the mmc4 dataset
    np.random.seed(42)
    full_indices_mmc4 = np.arange(len(rates_mmc4))
    selected_indices_mmc4 = np.random.choice(len(full_indices_mmc4), size=1742, replace=False)
    unseen_indices_mmc4 = np.setdiff1d(full_indices_mmc4, selected_indices_mmc4)
    # how many samples belong to HTCas9 at the front of hybrid_dataset
    offset = len(rates_HTCas9) + len(rates_HT11)
    # shift mmc4 indices into the global index space of hybrid_dataset
    unseen_indices_mmc4_global = (unseen_indices_mmc4 + offset).tolist()


    # Splitting the mmc5 dataset
    np.random.seed(42)
    full_indices_mmc5 = np.arange(len(rates_mmc5))
    selected_indices_mmc5 = np.random.choice(len(full_indices_mmc5), size=1835, replace=False)
    unseen_indices_mmc5 = np.setdiff1d(full_indices_mmc5, selected_indices_mmc5)
    # how many samples belong to HTCas9 at the front of hybrid_dataset
    offset = len(rates_HTCas9) + len(rates_HT11) + len(rates_mmc4)
    # shift mmc5 indices into the global index space of hybrid_dataset
    unseen_indices_mmc5_global = (unseen_indices_mmc5 + offset).tolist()

    # Splitting the change_seq dataset
    np.random.seed(42)
    full_indices_change_seq = np.arange(len(rates_change_seq))
    selected_indices_change_seq = np.random.choice(len(full_indices_change_seq), size=8000, replace=False)
    unseen_indices_change_seq = np.setdiff1d(full_indices_change_seq, selected_indices_change_seq)
    # how many samples belong to HTCas9 at the front of hybrid_dataset
    offset = len(rates_HTCas9) + len(rates_HT11) + len(rates_mmc4) + len(rates_mmc5)
    # shift change_seq indices into the global index space of hybrid_dataset
    unseen_indices_change_seq_global = (unseen_indices_change_seq + offset).tolist()

    # Splitting the tiger dataset
    np.random.seed(42)
    full_indices_tiger = np.arange(len(rates_tiger))
    selected_indices_tiger = np.random.choice(len(full_indices_tiger), size=8000, replace=False)
    unseen_indices_tiger = np.setdiff1d(full_indices_tiger, selected_indices_tiger)
    # how many samples belong to HTCas9 at the front of hybrid_dataset
    offset = len(rates_HTCas9) + len(rates_HT11) + len(rates_mmc4) + len(rates_mmc5) + len(rates_change_seq)
    # shift tiger indices into the global index space of hybrid_dataset
    unseen_indices_tiger_global = (unseen_indices_tiger + offset).tolist()

    # Splitting the iMeta dataset
    np.random.seed(42)
    full_indices_iMeta = np.arange(len(rates_iMeta))
    selected_indices_iMeta = np.random.choice(len(full_indices_iMeta), size=5776, replace=False)
    unseen_indices_iMeta = np.setdiff1d(full_indices_iMeta, selected_indices_iMeta)
    # how many samples belong to HTCas9 at the front of hybrid_dataset
    offset = len(rates_HTCas9) + len(rates_HT11) + len(rates_mmc4) + len(rates_mmc5) + len(rates_change_seq) + len(rates_tiger)
    # shift iMeta indices into the global index space of hybrid_dataset
    unseen_indices_iMeta_global = (unseen_indices_iMeta + offset).tolist()

    from torch.utils.data import ConcatDataset, DataLoader

    # Evaluate on 10 subsamples from unseen
    unseen_set_HTCas9 = Subset(hybrid_dataset, unseen_indices_HTCas9)
    unseen_set_HT11 = Subset(hybrid_dataset, unseen_indices_HT11_global)
    unseen_set_mmc4 = Subset(hybrid_dataset, unseen_indices_mmc4_global)
    unseen_set_mmc5 = Subset(hybrid_dataset, unseen_indices_mmc5_global)
    unseen_set_change_seq = Subset(hybrid_dataset, unseen_indices_change_seq_global)
    unseen_set_tiger = Subset(hybrid_dataset, unseen_indices_tiger_global)
    unseen_set_iMeta = Subset(hybrid_dataset, unseen_indices_iMeta_global)
    
    
    # Compute unseen Spearman for HTCas9 dataset
    spearman_list_HTCas9 = []
    for i in range(100):
        if len(unseen_set_HTCas9) < 1000:
            print("Not enough unseen samples to draw 1000 each time.")
            break
        idx_trial = np.random.choice(len(unseen_set_HTCas9), size=1000, replace=False)
        trial_subset = Subset(unseen_set_HTCas9, idx_trial)
        trial_loader = DataLoader(trial_subset, batch_size=1, shuffle=False, collate_fn=hybrid_collate)
        
        preds_trial = []
        labels_trial = []
        with torch.no_grad():
            for c_b, tbh_b, d_b, xx1_b, xx2_b, xx3_b, yy_b in trial_loader:
                p = model(c_b, tbh_b, d_b, xx1_b, xx2_b, xx3_b)
                preds_trial.append(p.item())
                labels_trial.append(yy_b.item())
        sp_corr, _ = spearmanr(labels_trial, preds_trial)
        spearman_list_HTCas9.append(sp_corr)
        print(sp_corr)
    avg_spearman = np.mean(spearman_list_HTCas9) if spearman_list_HTCas9 else None
    print(f"Average Spearman Correlation over 100 trials in HTCas9 dataset: {avg_spearman}")


    # Compute unseen Spearman for HT11 dataset
    spearman_list_HT11 = []
    for i in range(100):
        if len(unseen_set_HT11) < 1000:
            print("Not enough unseen samples to draw 1000 each time.")
            break
        idx_trial = np.random.choice(len(unseen_set_HT11), size=1000, replace=False)
        trial_subset = Subset(unseen_set_HT11, idx_trial)
        trial_loader = DataLoader(trial_subset, batch_size=1, shuffle=False, collate_fn=hybrid_collate)
        
        preds_trial = []
        labels_trial = []
        with torch.no_grad():
            for c_b, tbh_b, d_b, xx1_b, xx2_b, xx3_b, yy_b in trial_loader:
                p = model(c_b, tbh_b, d_b, xx1_b, xx2_b, xx3_b)
                preds_trial.append(p.item())
                labels_trial.append(yy_b.item())
        sp_corr, _ = spearmanr(labels_trial, preds_trial)
        spearman_list_HT11.append(sp_corr)

        print(sp_corr)
    avg_spearman = np.mean(spearman_list_HT11) if spearman_list_HT11 else None
    print(f"Average Spearman Correlation over 100 trials in HT11 dataset: {avg_spearman}")

    # Compute unseen Spearman for mmc4 dataset
    spearman_list_mmc4 = []
    for i in range(100):
        if len(unseen_set_mmc4) < 300:
            print("Not enough unseen samples to draw 300 each time.")
            break
        idx_trial = np.random.choice(len(unseen_set_mmc4), size=300, replace=False)
        trial_subset = Subset(unseen_set_mmc4, idx_trial)
        trial_loader = DataLoader(trial_subset, batch_size=1, shuffle=False, collate_fn=hybrid_collate)
        
        preds_trial = []
        labels_trial = []
        with torch.no_grad():
            for c_b, tbh_b, d_b, xx1_b, xx2_b, xx3_b, yy_b in trial_loader:
                p = model(c_b, tbh_b, d_b, xx1_b, xx2_b, xx3_b)
                preds_trial.append(p.item())
                labels_trial.append(yy_b.item())
        sp_corr, _ = spearmanr(labels_trial, preds_trial)
        spearman_list_mmc4.append(sp_corr)
        print(sp_corr)
    avg_spearman = np.mean(spearman_list_mmc4) if spearman_list_mmc4 else None
    print(f"Average Spearman Correlation over 100 trials in mmc4 dataset: {avg_spearman}")

    # Compute unseen Spearman for mmc5 dataset
    spearman_list_mmc5 = []
    for i in range(100):
        if len(unseen_set_mmc5) < 300:
            print("Not enough unseen samples to draw 300 each time.")
            break
        idx_trial = np.random.choice(len(unseen_set_mmc5), size=300, replace=False)
        trial_subset = Subset(unseen_set_mmc5, idx_trial)
        trial_loader = DataLoader(trial_subset, batch_size=1, shuffle=False, collate_fn=hybrid_collate)
        preds_trial = []
        labels_trial = []
        with torch.no_grad():
            for c_b, tbh_b, d_b, xx1_b, xx2_b, xx3_b, yy_b in trial_loader:
                p = model(c_b, tbh_b, d_b, xx1_b, xx2_b, xx3_b)
                preds_trial.append(p.item())
                labels_trial.append(yy_b.item())
        sp_corr, _ = spearmanr(labels_trial, preds_trial)
        spearman_list_mmc5.append(sp_corr)
        print(sp_corr)

    avg_spearman = np.mean(spearman_list_mmc5) if spearman_list_mmc5 else None
    print(f"Average Spearman Correlation over 100 trials in mmc5 dataset: {avg_spearman}")

    # Compute unseen Spearman for change_seq dataset
    spearman_list_change_seq = []
    for i in range(100):
        if len(unseen_set_change_seq) < 1000:
            print("Not enough unseen samples to draw 1000 each time.")
            break
        idx_trial = np.random.choice(len(unseen_set_change_seq), size=1000, replace=False)
        trial_subset = Subset(unseen_set_change_seq, idx_trial)
        trial_loader = DataLoader(trial_subset, batch_size=1, shuffle=False, collate_fn=hybrid_collate)
        preds_trial = []
        labels_trial = []
        with torch.no_grad():
            for c_b, tbh_b, d_b, xx1_b, xx2_b, xx3_b, yy_b in trial_loader:
                p = model(c_b, tbh_b, d_b, xx1_b, xx2_b, xx3_b)
                preds_trial.append(p.item())
                labels_trial.append(yy_b.item())
        sp_corr, _ = spearmanr(labels_trial, preds_trial)
        spearman_list_change_seq.append(sp_corr)
        print(sp_corr)

    avg_spearman = np.mean(spearman_list_change_seq) if spearman_list_change_seq else None
    print(f"Average Spearman Correlation over 100 trials in change_seq dataset: {avg_spearman}")

    # Compute unseen Spearman for tiger dataset
    spearman_list_tiger = []
    for i in range(100):
        if len(unseen_set_tiger) < 1000:
            print("Not enough unseen samples to draw 1000 each time.")
            break
        idx_trial = np.random.choice(len(unseen_set_tiger), size=1000, replace=False)
        trial_subset = Subset(unseen_set_tiger, idx_trial)
        trial_loader = DataLoader(trial_subset, batch_size=1, shuffle=False, collate_fn=hybrid_collate)
        preds_trial = []
        labels_trial = []
        with torch.no_grad():
            for c_b, tbh_b, d_b, xx1_b, xx2_b, xx3_b, yy_b in trial_loader:
                p = model(c_b, tbh_b, d_b, xx1_b, xx2_b, xx3_b)
                preds_trial.append(p.item())
                labels_trial.append(yy_b.item())
        sp_corr, _ = spearmanr(labels_trial, preds_trial)
        spearman_list_tiger.append(sp_corr)
        print(sp_corr)

    avg_spearman = np.mean(spearman_list_tiger) if spearman_list_tiger else None
    print(f"Average Spearman Correlation over 100 trials in tiger dataset: {avg_spearman}")

    # Compute unseen Spearman for iMeta dataset
    spearman_list_iMeta = []
    for i in range(100):
        if len(unseen_set_iMeta) < 1000:
            print("Not enough unseen samples to draw 1000 each time.")
            break
        idx_trial = np.random.choice(len(unseen_set_iMeta), size=1000, replace=False)
        trial_subset = Subset(unseen_set_iMeta, idx_trial)
        trial_loader = DataLoader(trial_subset, batch_size=1, shuffle=False, collate_fn=hybrid_collate)
        preds_trial = []
        labels_trial = []
        with torch.no_grad():
            for c_b, tbh_b, d_b, xx1_b, xx2_b, xx3_b, yy_b in trial_loader:
                p = model(c_b, tbh_b, d_b, xx1_b, xx2_b, xx3_b)
                preds_trial.append(p.item())
                labels_trial.append(yy_b.item())
        sp_corr, _ = spearmanr(labels_trial, preds_trial)
        spearman_list_iMeta.append(sp_corr)
        print(sp_corr)

    avg_spearman = np.mean(spearman_list_iMeta) if spearman_list_iMeta else None
    print(f"Average Spearman Correlation over 100 trials in iMeta dataset: {avg_spearman}")
    

if __name__ == "__main__":
    main_pipeline() 
```
