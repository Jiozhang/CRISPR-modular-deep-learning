```python
import os

ROOT = "19nt_spacer_datasets"
TARGET = "iMeta_combined_200_fine_tuned_preds"
NEW_NAME = "preds_200_fine_tuned_Generalist.txt"

renamed = 0
skipped = 0
errors = 0

for root, _, files in os.walk(ROOT):
    for fname in files:
        if not fname.lower().endswith(".txt"):
            continue

        # tolerate stray quotes in filenames
        fname_no_quotes = fname.replace('"', "").replace("'", "")

        if TARGET not in fname and TARGET not in fname_no_quotes:
            continue

        old_path = os.path.join(root, fname)
        new_path = os.path.join(root, NEW_NAME)

        # already correct name
        if os.path.abspath(old_path) == os.path.abspath(new_path):
            skipped += 1
            continue

        # do NOT overwrite existing Feature_CNN1_reduced.txt
        if os.path.exists(new_path):
            print(f"SKIP (exists): {new_path}  <-- would overwrite from {old_path}")
            skipped += 1
            continue

        try:
            os.rename(old_path, new_path)
            print(f"RENAMED: {old_path} -> {new_path}")
            renamed += 1
        except Exception as e:
            print(f"ERROR: {old_path} -> {new_path} | {e}")
            errors += 1

print("\nSummary")
print(f"Renamed: {renamed}")
print(f"Skipped: {skipped}")
print(f"Errors:  {errors}")
```

```python
import os
import re
import nbformat

ROOT = "19nt_spacer_datasets"

pattern = re.compile(r'["\'][^"\']*iMeta_combined_200_fine_tuned_preds[^"\']*\.txt["\']')
replacement = '"preds_200_fine_tuned_Generalist.txt"'

total_changes = 0
files_modified = 0

for root, _, files in os.walk(ROOT):
    for fn in files:
        if not fn.endswith(".ipynb"):
            continue

        path = os.path.join(root, fn)

        try:
            nb = nbformat.read(path, as_version=4)
        except Exception as e:
            print(f"Unreadable Notebook (skip): {path} | {e}")
            continue

        changes_in_file = 0
        for cell in nb.cells:
            if cell.get("cell_type") in ("code", "markdown"):
                new_source, n = pattern.subn(replacement, cell.get("source", ""))
                if n:
                    cell["source"] = new_source
                    changes_in_file += n

        if changes_in_file:
            nbformat.write(nb, path)
            print(f"Modified: {path} | Changes: {changes_in_file}")
            files_modified += 1
            total_changes += changes_in_file

print("\nSummary")
print(f"Files modified: {files_modified}")
print(f"Total replacements: {total_changes}")
```
