```python
# Sequence features in token

from nupack import *
import RNA
import math
import itertools

def DNA_reverse_complement(DNA):
    complement = {'A': 'T', 'C': 'G', 'G': 'C', 'T': 'A'}
    return ''.join(complement.get(base, base) for base in reversed(DNA))

def base_to_token(base):
    mapping = {
        "A": 0,
        "C": 1,
        "G": 2,
        "T": 3,
        "U": 3,
        "-": 0
    }
    return mapping.get(base, 0)

file_paths = [
    'Feature_CNN1_token.txt'
]

for file_path in file_paths:
    with open(file_path, 'w') as file:
        pass 

# Define NUPACK model
my_model_RNA = Model(material='rna', ensemble='nostacking', celsius=37, sodium=0.1, magnesium=0.01)
my_model_DNA = Model(material='dna', ensemble='nostacking', celsius=37, sodium=0.1, magnesium=0.01)

# Load crRNA sequences
file1 = open('iMeta_trans_LbCas12a_sorted_guide_TTTV_filtered.txt', 'r')    
lines = file1.readlines()
guide_array = [line.strip() for line in lines]

# Load target sequences
file2 = open('iMeta_trans_LbCas12a_sorted_target_full_TTTV_filtered.txt', 'r')    
lines = file2.readlines()
full_target_array = [line.strip() for line in lines]
file3 = open('iMeta_trans_LbCas12a_sorted_target_truncated_TTTV_filtered.txt', 'r')    
lines = file3.readlines()
truncated_target_array = [line.strip() for line in lines]

for i in range(len(guide_array)):

    # Initialize seq_array
    seq_array = []
    seq_tokens = []

    guide = guide_array[i]
    target_full = full_target_array[i]
    rc_target_full = DNA_reverse_complement(target_full)

    for base in guide:
        seq_tokens.append(base_to_token(base))

    for base in target_full:
        seq_tokens.append(base_to_token(base))

    for base in rc_target_full:
        seq_tokens.append(base_to_token(base))

    seq_array_to_append = [seq_tokens]
    seq_array.append(seq_array_to_append)

    seq_array = list(itertools.chain.from_iterable(seq_array))

    # Open a file in write mode
    with open('Feature_CNN1_token.txt', 'a') as file:
        # Iterate over each row in the 2D array
        for row in seq_array:
            # Convert each element to a string and join them with spaces
            file.write(' '.join(map(str, row)) + '\n')
        file.write('\n')

print('-------------')
```
