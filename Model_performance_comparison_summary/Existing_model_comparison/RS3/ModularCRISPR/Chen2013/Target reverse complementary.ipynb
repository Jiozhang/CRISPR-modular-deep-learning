```python
def reverse_complement(seq):
    complement = {
        'A': 'T', 'T': 'A',
        'C': 'G', 'G': 'C',
        'U': 'A',  # Optional: supports RNA input
        'N': 'N',
        'a': 't', 't': 'a',
        'c': 'g', 'g': 'c',
        'u': 'a',
        'n': 'n'
    }

    return ''.join(complement.get(base, base) for base in reversed(seq))


def reverse_complement_file(input_file, output_file):
    with open(input_file, "r") as infile, open(output_file, "w") as outfile:
        for line in infile:
            seq = line.strip()
            if seq:
                rc_seq = reverse_complement(seq)
                outfile.write(rc_seq + "\n")
            else:
                outfile.write("\n")  # Preserve empty lines


# Example usage
input_file = "train_target_20nt.txt"
output_file = "train_target_20nt_rc.txt"

reverse_complement_file(input_file, output_file)

print('finished')
```
