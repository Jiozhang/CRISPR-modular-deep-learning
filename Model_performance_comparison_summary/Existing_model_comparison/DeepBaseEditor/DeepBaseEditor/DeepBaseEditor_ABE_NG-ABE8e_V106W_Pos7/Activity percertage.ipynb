```python
input_file = "activity.txt"

# Read, divide by 100, and overwrite the file
with open(input_file, "r") as f:
    lines = f.readlines()

with open(input_file, "w") as f:
    for line in lines:
        line = line.strip()
        if not line:  # Skip empty lines
            continue
        value = float(line)
        f.write(f"{value / 100}\n")

print("Finished dividing all values by 100.")
```
