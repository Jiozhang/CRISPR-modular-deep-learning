```python
from pathlib import Path

# Reference file
reference_file = "activity.txt"

# Other files whose corresponding lines should also be removed
other_files = [
    "guide_seq.txt",
    "target_context_seq.txt",
    "pam.txt"
]

# Read all files
with open(reference_file, "r") as f:
    ref_lines = f.readlines()

other_lines = []
for file in other_files:
    with open(file, "r") as f:
        other_lines.append(f.readlines())

# Check that all files have the same number of lines
num_lines = len(ref_lines)
for i, lines in enumerate(other_lines):
    if len(lines) != num_lines:
        raise ValueError(
            f"{other_files[i]} has {len(lines)} lines, "
            f"but {reference_file} has {num_lines} lines."
        )

# Find indices to keep
keep_indices = [
    i for i, line in enumerate(ref_lines)
    if line.strip().lower() != "na"
]

# Write filtered reference file
with open(reference_file, "w") as f:
    for i in keep_indices:
        f.write(ref_lines[i])

# Write filtered other files
for file, lines in zip(other_files, other_lines):
    with open(file, "w") as f:
        for i in keep_indices:
            f.write(lines[i])

print(f"Removed {num_lines - len(keep_indices)} rows.")
```
