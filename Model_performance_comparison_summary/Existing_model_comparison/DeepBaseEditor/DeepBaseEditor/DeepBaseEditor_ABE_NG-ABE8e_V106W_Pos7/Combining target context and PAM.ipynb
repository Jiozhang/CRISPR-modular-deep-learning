```python
target_file = "target_context_seq.txt"
pam_file = "pam.txt"
output_file = "target_context_30nt.txt"

# Read all lines
with open(target_file, "r") as f:
    target_lines = [line.strip() for line in f]

with open(pam_file, "r") as f:
    pam_lines = [line.strip() for line in f]

# Check that both files have the same number of lines
if len(target_lines) != len(pam_lines):
    raise ValueError(
        f"The two files have different numbers of lines: "
        f"{len(target_lines)} vs {len(pam_lines)}"
    )

# Concatenate corresponding lines and write to output
with open(output_file, "w") as f:
    for target, pam in zip(target_lines, pam_lines):
        f.write(target + pam + "\n")

print(f"Concatenated sequences saved to '{output_file}'.")
```
