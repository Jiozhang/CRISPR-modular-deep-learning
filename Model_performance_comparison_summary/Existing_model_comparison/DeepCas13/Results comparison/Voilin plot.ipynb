```python
import numpy as np
import matplotlib.pyplot as plt

# -----------------------------
# Load data
# -----------------------------
group1 = np.loadtxt("Modular_full_Validation_loss.txt")
group2 = np.loadtxt("Benchmark_Validation_loss.txt")

# Ignore NaN values
group1 = group1[~np.isnan(group1)]
group2 = group2[~np.isnan(group2)]

# -----------------------------
# Normalize using global min/max
# -----------------------------
all_values = np.concatenate([group1, group2])
vmin = all_values.min()
vmax = all_values.max()

group1 = (group1 - vmin) / (vmax - vmin)
group2 = (group2 - vmin) / (vmax - vmin)

# -----------------------------
# Median difference
# -----------------------------
median_group1 = np.median(group1)
median_group2 = np.median(group2)

median_difference = median_group1 - median_group2

print(f"Median difference: {median_difference:.6f}")

# -----------------------------
# Plot settings
# -----------------------------
plt.rcParams.update({
    "font.family": "Arial",
    "font.size": 20,
    "axes.linewidth": 1.5,
    "xtick.major.width": 1.5,
    "ytick.major.width": 1.5,
})

fig, ax = plt.subplots(figsize=(5.5, 6), dpi=600)

# Violin plot
parts = ax.violinplot(
    [group1, group2],
    positions=[1, 2],
    widths=0.7,
    showmeans=False,
    showmedians=True,
    showextrema=False,
    quantiles=[[0.25, 0.75], [0.25, 0.75]]
)

# Style violins
for body in parts['bodies']:
    body.set_facecolor("#4C72B0")
    body.set_edgecolor("black")
    body.set_alpha(0.35)
    body.set_linewidth(1.5)

parts['cmedians'].set_color("black")
parts['cmedians'].set_linewidth(2)

parts['cquantiles'].set_color("black")
parts['cquantiles'].set_linestyle("--")
parts['cquantiles'].set_linewidth(1.2)

# -----------------------------
# Overlay individual points
# -----------------------------
rng = np.random.default_rng(42)

for i, data in enumerate([group1, group2], start=1):
    x = rng.normal(i, 0.04, len(data))
    ax.scatter(
        x,
        data,
        s=20,
        color="black",
        alpha=0.45,
        zorder=3
    )

# -----------------------------
# Labels
# -----------------------------
ax.set_xticks([1, 2])
ax.set_xticklabels(["ModularCRISPR", "DeepCas13"])

ax.set_ylabel("Normalized Validation Loss", fontsize=24)
ax.set_ylim(-0.02, 1.02)

# Remove unnecessary borders
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

# Light horizontal grid
ax.grid(axis="y", linestyle="--", alpha=0.3)

plt.tight_layout()
plt.savefig("Validation_loss_violin_plot.png", dpi=600, bbox_inches="tight")
plt.show()
```
