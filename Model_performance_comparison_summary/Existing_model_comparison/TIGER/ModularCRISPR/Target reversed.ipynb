```python
def front_join_reverse_and_convert_to_rna(target_file, context_file, output_file):
    with open(target_file, "r") as target_in, \
         open(context_file, "r") as context_in, \
         open(output_file, "w") as outfile:

        for target_line, context_line in zip(target_in, context_in):
            target = target_line.strip()
            context = context_line.strip()

            combined_sequence = context + target
            reversed_sequence = combined_sequence[::-1]
            rna_sequence = reversed_sequence.replace("T", "U").replace("t", "u")

            outfile.write(rna_sequence + "\n")


front_join_reverse_and_convert_to_rna(
    "target_sequences.txt",
    "5p_context.txt",
    "full_target_sequences.txt"
)

print("finished")
```

```python
def reverse_and_convert_to_rna(target_file, output_file):
    with open(target_file, "r") as target_in, \
         open(output_file, "w") as outfile:

        for target_line in target_in:
            target = target_line.strip()
            reversed_sequence = target[::-1]
            rna_sequence = reversed_sequence.replace("T", "U").replace("t", "u")
            outfile.write(rna_sequence + "\n")


target_file = "target_sequences.txt"
output_file = "target_sequences_r.txt"

reverse_and_convert_to_rna(target_file, output_file)

print("finished")
```
