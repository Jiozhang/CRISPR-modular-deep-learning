```python
def reverse_complement(seq):
    complement = {
        'A': 'T', 'T': 'A',
        'C': 'G', 'G': 'C',
        'U': 'A',  # Optional: supports RNA input
        'N': 'N',
        'a': 't', 't': 'a',
        'c': 'g', 'g': 'c',
        'u': 'a',
        'n': 'n'
    }

    return ''.join(complement.get(base, base) for base in reversed(seq))


def reverse_complement_and_extract(input_file, output_file):
    with open(input_file, "r") as infile, open(output_file, "w") as outfile:
        for line in infile:
            seq = line.strip()

            if seq:
                # Reverse complement
                rc_seq = reverse_complement(seq)

                # Extract 4th–37th bases (1-based indexing)
                context_34nt = rc_seq[6:40]

                outfile.write(context_34nt + "\n")
            else:
                outfile.write("\n")  # Preserve empty lines


# Example usage
input_file = "HT1-1_target_sequences_with_noise_rc.txt"
output_file = "target_context_34nt.txt"

reverse_complement_and_extract(input_file, output_file)

print("Finished")
```
