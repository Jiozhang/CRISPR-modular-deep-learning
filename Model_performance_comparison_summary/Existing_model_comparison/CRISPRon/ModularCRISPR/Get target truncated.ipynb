```python
input_file = "train_30mer_gRNA.txt"
output_file = "train_20mer_gRNA.txt"

with open(input_file, "r") as infile, open(output_file, "w") as outfile:
    for line in infile:
        sequence = line.strip()

        # Extract 5th through 24th symbols (inclusive)
        extracted = sequence[4:24]

        outfile.write(extracted + "\n")

print(f"Done! Output saved to: {output_file}")
```
