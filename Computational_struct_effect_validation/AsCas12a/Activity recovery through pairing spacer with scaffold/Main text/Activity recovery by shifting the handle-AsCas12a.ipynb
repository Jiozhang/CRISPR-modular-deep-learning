```python
import RNA
from nupack import *
import random
import math

rate_a = []
rate_b = []
rate_c = []


with open('test_generated_random_dna_5_pairs_18_22_paired_with_28_32_preds_HT11.txt', 'r') as f:
    rate_a = [line.strip() for line in f]

with open('test_generated_random_dna_5_pairs_20_24_paired_with_30_34_preds_HT11.txt', 'r') as f:
    rate_b = [line.strip() for line in f]

with open('test_generated_random_dna_5_pairs_21_25_paired_with_31_35_preds_HT11.txt', 'r') as f:
    rate_c = [line.strip() for line in f]


import statistics

median_a = statistics.median(map(float, rate_a))
median_b = statistics.median(map(float, rate_b))
median_c = statistics.median(map(float, rate_c))


'''
average_p = sum(map(float, rate_p)) / len(rate_p)
average_n = sum(map(float, rate_n)) / len(rate_n)
'''

print(median_a, median_b, median_c)


import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import statistics

# Organize the rate lists into a dictionary
score_bin_data = {
    '18-22\nwith\n28-32': rate_a,
    '20-24\nwith\n30-34': rate_b,
    '21-25\nwith\n31-35': rate_c
}

# Convert to DataFrame
plot_data = []
for label, rates in score_bin_data.items():
    for r in rates:
        try:
            plot_data.append({'Group': label, 'Predicted Indel Frequency (%)': float(r)})
        except:
            continue  # skip invalid data points

df = pd.DataFrame(plot_data)
counts = df.groupby("Group")["Predicted Indel Frequency (%)"].count()
print(counts)

# Plot
plt.figure(figsize=(4, 6), dpi=600)

sns.boxplot(
    data=df,
    x='Group',
    y='Predicted Indel Frequency (%)',
    palette='coolwarm_r',
    # Box outline
    linewidth=2.5,
    boxprops=dict(edgecolor='black'),
    # Whiskers (vertical)
    whiskerprops=dict(color='black', linewidth=2.5),
    # Caps (horizontal)
    capprops=dict(color='black', linewidth=2.5),
    # Median line
    medianprops=dict(color='black', linewidth=3),
    # Outliers (optional)
    flierprops=dict(
        marker='o',
        markersize=6,
        markerfacecolor='none',   # <-- hollow
        markeredgecolor='black',
        linewidth=1.5,
        alpha=0.8
    )
)

from matplotlib.ticker import MaxNLocator

ax = plt.gca()

ax.yaxis.set_major_locator(MaxNLocator(nbins=5))

plt.ylabel("Predicted Activity", fontsize=24)
plt.xlabel("")          # removes the x-axis title
plt.xticks(fontsize=18)
plt.yticks(fontsize=24)
plt.tight_layout()
plt.show()
```
