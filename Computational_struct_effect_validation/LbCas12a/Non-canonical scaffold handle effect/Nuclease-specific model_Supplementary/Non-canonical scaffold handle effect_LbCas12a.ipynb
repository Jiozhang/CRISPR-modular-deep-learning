```python
import RNA
from nupack import *
import random
import math

rate_a = []
rate_b = []
rate_c = []
rate_d = []
rate_e = []
rate_f = []
rate_g = []

with open('test_generated_random_dna_linear_preds_iMeta.txt', 'r') as f:
    rate_a = [line.strip() for line in f]

with open('test_generated_random_dna_3_pairs_consecutive_preds_iMeta.txt', 'r') as f:
    rate_b = [line.strip() for line in f]

with open('test_generated_random_dna_4_pairs_consecutive_preds_iMeta.txt', 'r') as f:
    rate_c = [line.strip() for line in f]

with open('test_generated_random_dna_5_pairs_consecutive_preds_iMeta.txt', 'r') as f:
    rate_d = [line.strip() for line in f]

with open('test_generated_random_dna_6_pairs_consecutive_preds_iMeta.txt', 'r') as f:
    rate_e = [line.strip() for line in f]

with open('test_generated_random_dna_7_pairs_consecutive_preds_iMeta.txt', 'r') as f:
    rate_f = [line.strip() for line in f]

with open('test_generated_random_dna_8_pairs_consecutive_preds_iMeta.txt', 'r') as f:
    rate_g = [line.strip() for line in f]

    
import statistics

median_a = statistics.median(map(float, rate_a))
median_b = statistics.median(map(float, rate_b))
median_c = statistics.median(map(float, rate_c))
median_d = statistics.median(map(float, rate_d))
median_e = statistics.median(map(float, rate_e))
median_f = statistics.median(map(float, rate_f))
median_g = statistics.median(map(float, rate_g))


'''
average_p = sum(map(float, rate_p)) / len(rate_p)
average_n = sum(map(float, rate_n)) / len(rate_n)
'''

print(median_a, median_b, median_c, median_d, median_e, median_f, median_g)


import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import statistics

# Organize the rate lists into a dictionary
score_bin_data = {
    'None': rate_a,
    '3 pairs': rate_b,
    '4 pairs': rate_c,
    '5 pairs': rate_d,
    '6 pairs': rate_e,
    '7 pairs': rate_f,
    '8 pairs': rate_g
}


# Convert to DataFrame
plot_data = []
for label, rates in score_bin_data.items():
    for r in rates:
        try:
            plot_data.append({'Group': label, 'Predicted Indel Frequency': float(r)})
        except:
            continue  # skip invalid data points

df = pd.DataFrame(plot_data)
counts = df.groupby("Group")["Predicted Indel Frequency"].count()
print(counts)

group_order = ['None', '3 pairs', '4 pairs', '5 pairs', '6 pairs', '7 pairs', '8 pairs']
medians = df.groupby('Group')['Predicted Indel Frequency'].median()

y1 = medians['None']
y2 = medians['8 pairs']
delta = y2 - y1
print(delta)

# Plot
plt.figure(figsize=(12, 10), dpi=600)  # higher DPI for sharper rendering
ax = sns.boxplot(
    data=df,
    x='Group',
    y='Predicted Indel Frequency',
    palette='coolwarm_r',
    # Box outline
    linewidth=2.5,
    boxprops=dict(edgecolor='black'),
    # Whiskers (vertical)
    whiskerprops=dict(color='black', linewidth=2.5),
    # Caps (horizontal)
    capprops=dict(color='black', linewidth=2.5),
    # Median line
    medianprops=dict(color='black', linewidth=3),
    # Outliers (optional)
    flierprops=dict(
        marker='o',
        markersize=6,
        markerfacecolor='none',   # <-- hollow
        markeredgecolor='black',
        linewidth=1.5,
        alpha=0.8
    )
)
# x positions (0-based index)
x1, x2 = 0, len(group_order) - 1

# height for the bracket
y_max = df['Predicted Indel Frequency'].max()
h = (df['Predicted Indel Frequency'].max() -
     df['Predicted Indel Frequency'].min()) * 0.05
y = y_max + h

# draw bracket
ax.plot([x1, x1, x2, x2], [y, y + h, y + h, y], lw=2, c='black')

# add text
ax.text(
    (x1 + x2) * 0.5,
    y + h * 1.2,
    f'Δ median = {delta:.3f}',
    ha='center',
    va='bottom',
    fontsize=23
)


# Optional: add raw data overlay (commented out)
# sns.stripplot(data=df, x='Group', y='Indel Frequency (%)', color='black', alpha=0.4, jitter=0.2, size=3)

plt.title("LbCas12a Canonical Scaffold", fontsize=32)
plt.ylabel("Predicted activity", fontsize=32)
plt.xlabel("Number of bases paired in spacer", fontsize=32)
plt.xticks(fontsize=24)
plt.yticks(fontsize=32)
plt.tight_layout()
plt.show()
```
