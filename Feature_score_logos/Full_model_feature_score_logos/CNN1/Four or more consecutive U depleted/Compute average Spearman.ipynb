```python
import pandas as pd

# Read CSV file
df = pd.read_csv("SpCas9_computed_spearman_by_base_position.csv")

# Ensure correct data types
df["Position"] = pd.to_numeric(df["Position"])
df["Computed_Spearman"] = pd.to_numeric(df["Computed_Spearman"])

# Define windows
windows = {
    "First_6": (1, 6),
    "Middle_8": (7, 14),
    "Last_6": (15, 20)
}

bases = ["A", "C", "G", "T"]

results = []

for base in bases:
    base_df = df[df["Base"] == base]

    row = {"Base": base}

    for window_name, (start, end) in windows.items():
        avg = base_df.loc[
            base_df["Position"].between(start, end),
            "Computed_Spearman"
        ].mean()

        row[window_name] = avg

    results.append(row)

# Convert to DataFrame
result_df = pd.DataFrame(results)

print(result_df)
```
