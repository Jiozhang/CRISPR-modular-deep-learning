```python
import pandas as pd

# Read CSV file
df = pd.read_csv("SpCas9_computed_spearman_by_struct_position_ssDNA_target_ah.csv")

# Ensure correct data types
df["Position"] = pd.to_numeric(df["Position"])
df["Computed_Spearman"] = pd.to_numeric(df["Computed_Spearman"])

# Define windows
windows = {
    "First_6": (1, 6),
    "Middle_8": (7, 14),
    "Last_6": (15, 20)
}

structs = ["U", "O", "C"]

results = []

for struct in structs:
    struct_df = df[df["Struct"] == struct]

    row = {"Struct": struct}

    for window_name, (start, end) in windows.items():
        avg = struct_df.loc[
            struct_df["Position"].between(start, end),
            "Computed_Spearman"
        ].mean()

        row[window_name] = avg

    results.append(row)

# Convert to DataFrame
result_df = pd.DataFrame(results)

print(result_df)
```
