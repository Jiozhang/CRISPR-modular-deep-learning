```python
import random


# ============================================================
# Input files
# ============================================================

GUIDE_FILE = (
    "decreased_relative_activity_filtered_full_guide_seqs.txt"
)

TARGET_FILE = (
    "decreased_relative_activity_filtered_target_seqs_rc.txt"
)

ACTIVITY_FILE = (
    "decreased_relative_activity_filtered_log_transformed_activity.txt"
)


# ============================================================
# Output files
# ============================================================

OUTPUT_GUIDE_FILE = (
    "decreased_relative_activity_filtered_full_guide_seqs_subsampled_1.txt"
)

OUTPUT_TARGET_FILE = (
    "decreased_relative_activity_filtered_target_seqs_rc_subsampled_1.txt"
)

OUTPUT_ACTIVITY_FILE = (
    "sampled_10000_decreased_relative_activity_filtered_log_transformed_activity_subsampled_1.txt"
)


# ============================================================
# Settings
# ============================================================

SAMPLE_SIZE = 10000
RANDOM_SEED = 42


def read_lines(file_path):
    """
    Read all lines while preserving their original content,
    except for the trailing newline character.
    """
    with open(file_path, "r") as file:
        return [line.rstrip("\r\n") for line in file]


# ============================================================
# 1. Read the three files
# ============================================================

guide_lines = read_lines(GUIDE_FILE)
target_lines = read_lines(TARGET_FILE)
activity_lines = read_lines(ACTIVITY_FILE)


# ============================================================
# 2. Confirm that all files have the same number of lines
# ============================================================

if not (
    len(guide_lines)
    == len(target_lines)
    == len(activity_lines)
):
    raise ValueError(
        "The three files do not contain the same number of lines.\n"
        f"Guide file:    {len(guide_lines)}\n"
        f"Target file:   {len(target_lines)}\n"
        f"Activity file: {len(activity_lines)}"
    )


total_samples = len(guide_lines)

if total_samples < SAMPLE_SIZE:
    raise ValueError(
        f"Only {total_samples} samples are available, but "
        f"{SAMPLE_SIZE} samples were requested."
    )


# ============================================================
# 3. Randomly select 10,000 shared line indices
# ============================================================

random.seed(RANDOM_SEED)

sampled_indices = random.sample(
    range(total_samples),
    SAMPLE_SIZE
)

# Sort the indices so the selected samples remain in their
# original order in the output files.
sampled_indices.sort()


# ============================================================
# 4. Write the same selected lines to all three output files
# ============================================================

with open(OUTPUT_GUIDE_FILE, "w") as guide_output, \
     open(OUTPUT_TARGET_FILE, "w") as target_output, \
     open(OUTPUT_ACTIVITY_FILE, "w") as activity_output:

    for index in sampled_indices:
        guide_output.write(guide_lines[index] + "\n")
        target_output.write(target_lines[index] + "\n")
        activity_output.write(activity_lines[index] + "\n")


# ============================================================
# 5. Optional: save the original selected line numbers
# ============================================================

INDEX_OUTPUT_FILE = "sampled_10000_original_line_numbers.txt"

with open(INDEX_OUTPUT_FILE, "w") as index_output:
    for index in sampled_indices:
        # Add 1 because human-readable line numbers start from 1
        index_output.write(f"{index + 1}\n")


# ============================================================
# 6. Print summary
# ============================================================

print("Sampling completed.")
print(f"Random seed:      {RANDOM_SEED}")
print(f"Original samples: {total_samples}")
print(f"Sampled samples:  {SAMPLE_SIZE}")
print()

print("Output files:")
print(f"  {OUTPUT_GUIDE_FILE}")
print(f"  {OUTPUT_TARGET_FILE}")
print(f"  {OUTPUT_ACTIVITY_FILE}")
print(f"  {INDEX_OUTPUT_FILE}")
```
