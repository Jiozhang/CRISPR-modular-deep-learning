```python
import math

input_file = "raw_activity.txt"
output_file = "log_transformed_activity.txt"

with open(input_file, "r") as infile, open(output_file, "w") as outfile:
    for line_number, line in enumerate(infile, start=1):
        line = line.strip()

        # Skip empty lines
        if not line:
            continue

        try:
            x = float(line)

            if x < -1:
                raise ValueError(
                    f"log10(x+1) is undefined because x = {x} < -1."
                )

            y = math.log10(x + 1)

            outfile.write(f"{y:.10g}\n")

        except ValueError as e:
            print(f"Skipping line {line_number}: {e}")

print(f"Log-transformed values saved to '{output_file}'.")
```
